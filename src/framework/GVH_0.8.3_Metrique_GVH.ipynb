### Variabilité de la norme de transition

Soit une série de \(N_T\) normes de transition

\[
D_T(1),D_T(2),\ldots,D_T(N_T).
\]

La norme moyenne de transition est définie par

\[
\overline{D_T}
=
\frac{1}{N_T}
\sum_{k=1}^{N_T}D_T(k).
\]

L'observable de variabilité absolue est définie par

\[
\boxed{
S_T
=
\sqrt{
\frac{1}{N_T}
\sum_{k=1}^{N_T}
\left(
D_T(k)-\overline{D_T}
\right)^2
}
}
\]

Elle correspond à l'écart-type population de la série \(D_T(k)\).

Afin de comparer des systèmes dont les amplitudes moyennes de
transition sont différentes, on introduit la variabilité relative

\[
\boxed{
S_T^{\mathrm{norm}}
=
\frac{S_T}{\overline{D_T}}
}
\]

lorsque \(\overline{D_T}>0\).

\(S_T\) mesure la dispersion absolue des amplitudes de transition,
tandis que \(S_T^{\mathrm{norm}}\) mesure cette dispersion relativement
à leur amplitude moyenne.

À ce stade, ces observables décrivent uniquement la variabilité de
\(D_T\). Elles ne constituent pas, à elles seules, une mesure de chaos,
d'instabilité ou de perturbation physique.

```python
# ============================================================
# Robustesse de D_T, S_T et S_T_norm au pas temporel
# Normalisation fixe — seuls les intervalles temporels changent
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ------------------------------------------------------------
# 1. Pas temporels étudiés
# ------------------------------------------------------------

# Les données primitives sont horaires.
pas_heures = [1, 2, 3, 6, 12, 24]


# ------------------------------------------------------------
# 2. Contrôles préalables
# ------------------------------------------------------------

X_normalise = np.asarray(
    X_normalise,
    dtype=float
)

if X_normalise.ndim != 2:
    raise ValueError(
        "X_normalise doit être une matrice bidimensionnelle."
    )

if X_normalise.shape[1] != 18:
    raise ValueError(
        "L'état collectif normalisé doit contenir 18 composantes."
    )

if not np.isfinite(X_normalise).all():
    raise ValueError(
        "X_normalise contient des valeurs non finies."
    )

n_etats = X_normalise.shape[0]

print("Nombre d'états normalisés :", n_etats)
print("Dimension de l'état collectif :", X_normalise.shape[1])


# ------------------------------------------------------------
# 3. Fonction de calcul pour un pas temporel donné
# ------------------------------------------------------------

def calculer_observables_pas_temporel(
    X_norm,
    lag,
    pas_elementaire_heures=1.0
):
    """
    Calcule D_T, S_T et S_T_norm pour un décalage temporel donné.

    Paramètres
    ----------
    X_norm : ndarray, forme (N, 18)
        États collectifs déjà normalisés.

    lag : int
        Nombre de pas élémentaires séparant deux états comparés.

    pas_elementaire_heures : float
        Durée du pas primitif en heures.

    Retour
    ------
    dict
        Séries D_T collectives et individuelles,
        ainsi que leurs statistiques.
    """

    if not isinstance(lag, (int, np.integer)):
        raise TypeError(
            "lag doit être un entier."
        )

    if lag <= 0:
        raise ValueError(
            "lag doit être strictement positif."
        )

    if lag >= X_norm.shape[0]:
        raise ValueError(
            "lag est trop grand par rapport au nombre d'états."
        )

    # Transition entre X_k et X_{k+lag}
    Delta_lag = (
        X_norm[lag:, :]
        -
        X_norm[:-lag, :]
    )

    # Décomposition selon l'ordre déjà adopté :
    # colonnes 0:6   -> Io
    # colonnes 6:12  -> Europa
    # colonnes 12:18 -> Ganymede

    Delta_Io = Delta_lag[:, 0:6]
    Delta_Europa = Delta_lag[:, 6:12]
    Delta_Ganymede = Delta_lag[:, 12:18]

    # Normes de transition
    D_collectif = np.linalg.norm(
        Delta_lag,
        axis=1
    )

    D_Io = np.linalg.norm(
        Delta_Io,
        axis=1
    )

    D_Europa = np.linalg.norm(
        Delta_Europa,
        axis=1
    )

    D_Ganymede = np.linalg.norm(
        Delta_Ganymede,
        axis=1
    )

    # Vérification de la décomposition quadratique
    reconstruction = np.sqrt(
        D_Io**2
        +
        D_Europa**2
        +
        D_Ganymede**2
    )

    erreur_decomposition = np.max(
        np.abs(
            D_collectif
            -
            reconstruction
        )
    )

    decomposition_validee = np.allclose(
        D_collectif,
        reconstruction,
        rtol=1e-13,
        atol=1e-13
    )

    def statistiques(D):
        moyenne = np.mean(D)
        ecart_type = np.std(
            D,
            ddof=0
        )

        if np.isclose(moyenne, 0.0):
            ecart_type_norm = np.nan
        else:
            ecart_type_norm = (
                ecart_type
                /
                moyenne
            )

        return {
            "N_transitions": D.size,
            "D_T_mean": moyenne,
            "S_T": ecart_type,
            "S_T_norm": ecart_type_norm,
            "S_T_norm_percent": 100.0 * ecart_type_norm,
            "D_T_min": np.min(D),
            "D_T_median": np.median(D),
            "D_T_max": np.max(D)
        }

    return {
        "lag": lag,
        "Delta_t_heures": lag * pas_elementaire_heures,
        "D_T_collectif": D_collectif,
        "D_T_Io": D_Io,
        "D_T_Europa": D_Europa,
        "D_T_Ganymede": D_Ganymede,
        "stats_collectif": statistiques(D_collectif),
        "stats_Io": statistiques(D_Io),
        "stats_Europa": statistiques(D_Europa),
        "stats_Ganymede": statistiques(D_Ganymede),
        "decomposition_validee": decomposition_validee,
        "erreur_decomposition": erreur_decomposition
    }


# ------------------------------------------------------------
# 4. Calcul pour tous les pas temporels
# ------------------------------------------------------------

resultats_multi_pas = {}

for lag in pas_heures:

    resultat = calculer_observables_pas_temporel(
        X_norm=X_normalise,
        lag=lag,
        pas_elementaire_heures=1.0
    )

    resultats_multi_pas[lag] = resultat


# ------------------------------------------------------------
# 5. Construction du tableau collectif
# ------------------------------------------------------------

lignes_collectif = []

for lag, resultat in resultats_multi_pas.items():

    stats = resultat["stats_collectif"]

    lignes_collectif.append({
        "Delta_t_heures": resultat["Delta_t_heures"],
        "N_transitions": stats["N_transitions"],
        "D_T_mean": stats["D_T_mean"],
        "S_T": stats["S_T"],
        "S_T_norm": stats["S_T_norm"],
        "S_T_norm_percent": stats["S_T_norm_percent"],
        "D_T_min": stats["D_T_min"],
        "D_T_median": stats["D_T_median"],
        "D_T_max": stats["D_T_max"],
        "D_T_mean_par_heure":
            stats["D_T_mean"]
            /
            resultat["Delta_t_heures"],
        "S_T_par_heure":
            stats["S_T"]
            /
            resultat["Delta_t_heures"],
        "Decomposition_validee":
            resultat["decomposition_validee"],
        "Erreur_decomposition":
            resultat["erreur_decomposition"]
    })

resume_multi_pas_collectif = pd.DataFrame(
    lignes_collectif
)

print(
    "\nRobustesse temporelle — système collectif :"
)

display(
    resume_multi_pas_collectif
)


# ------------------------------------------------------------
# 6. Tableau détaillé collectif + satellites
# ------------------------------------------------------------

lignes_detaillees = []

for lag, resultat in resultats_multi_pas.items():

    groupes = {
        "Collectif": resultat["stats_collectif"],
        "Io": resultat["stats_Io"],
        "Europa": resultat["stats_Europa"],
        "Ganymede": resultat["stats_Ganymede"]
    }

    for systeme, stats in groupes.items():

        lignes_detaillees.append({
            "Systeme": systeme,
            "Delta_t_heures":
                resultat["Delta_t_heures"],
            "N_transitions":
                stats["N_transitions"],
            "D_T_mean":
                stats["D_T_mean"],
            "S_T":
                stats["S_T"],
            "S_T_norm":
                stats["S_T_norm"],
            "S_T_norm_percent":
                stats["S_T_norm_percent"],
            "D_T_mean_par_heure":
                stats["D_T_mean"]
                /
                resultat["Delta_t_heures"]
        })

resume_multi_pas_detaille = pd.DataFrame(
    lignes_detaillees
)

print(
    "\nRobustesse temporelle — collectif et satellites :"
)

display(
    resume_multi_pas_detaille
)


# ------------------------------------------------------------
# 7. Vérification du cas de référence Delta t = 1 heure
# ------------------------------------------------------------

resultat_1h = resultats_multi_pas[1]

verification_D_T_1h = np.allclose(
    resultat_1h["D_T_collectif"],
    D_T,
    rtol=1e-14,
    atol=1e-14
)

erreur_D_T_1h = np.max(
    np.abs(
        resultat_1h["D_T_collectif"]
        -
        D_T
    )
)

print(
    "\nVérification du cas Δt = 1 heure :"
)

print(
    "D_T multi-pas identique au D_T initial :",
    verification_D_T_1h
)

print(
    "Erreur maximale :",
    f"{erreur_D_T_1h:.3e}"
)


# ------------------------------------------------------------
# 8. Test de proportionnalité locale D_T ~ Delta t
# ------------------------------------------------------------

D_mean_1h = (
    resultat_1h["stats_collectif"]["D_T_mean"]
)

lignes_scaling = []

for lag, resultat in resultats_multi_pas.items():

    delta_t = resultat["Delta_t_heures"]

    D_mean_observe = (
        resultat["stats_collectif"]["D_T_mean"]
    )

    # Valeur attendue sous croissance strictement linéaire
    D_mean_lineaire = (
        D_mean_1h
        *
        delta_t
    )

    ratio_scaling = (
        D_mean_observe
        /
        D_mean_lineaire
    )

    ecart_relatif = (
        ratio_scaling
        -
        1.0
    )

    lignes_scaling.append({
        "Delta_t_heures": delta_t,
        "D_T_mean_observe": D_mean_observe,
        "D_T_mean_lineaire": D_mean_lineaire,
        "Ratio_observe_lineaire": ratio_scaling,
        "Ecart_relatif": ecart_relatif,
        "Ecart_relatif_percent":
            100.0 * ecart_relatif
    })

resume_scaling = pd.DataFrame(
    lignes_scaling
)

print(
    "\nTest de croissance locale D_T ∝ Δt :"
)

display(
    resume_scaling
)


# ------------------------------------------------------------
# 9. Graphique : moyenne de D_T selon le pas temporel
# ------------------------------------------------------------

plt.figure(
    figsize=(9, 5)
)

plt.plot(
    resume_multi_pas_collectif["Delta_t_heures"],
    resume_multi_pas_collectif["D_T_mean"],
    marker="o",
    label="D_T moyen observé"
)

plt.plot(
    resume_scaling["Delta_t_heures"],
    resume_scaling["D_T_mean_lineaire"],
    marker="s",
    linestyle="--",
    label="Référence linéaire issue de 1 h"
)

plt.xlabel(
    "Pas temporel Δt (heures)"
)

plt.ylabel(
    "D_T moyen"
)

plt.title(
    "Robustesse de D_T au changement du pas temporel"
)

plt.grid(
    alpha=0.3
)

plt.legend()

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 10. Graphique : S_T_norm selon le pas temporel
# ------------------------------------------------------------

plt.figure(
    figsize=(9, 5)
)

plt.plot(
    resume_multi_pas_collectif["Delta_t_heures"],
    resume_multi_pas_collectif["S_T_norm_percent"],
    marker="o"
)

plt.xlabel(
    "Pas temporel Δt (heures)"
)

plt.ylabel(
    "S_T_norm (%)"
)

plt.title(
    "Variabilité relative des transitions selon le pas temporel"
)

plt.grid(
    alpha=0.3
)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 11. Graphique comparatif des satellites
# ------------------------------------------------------------

plt.figure(
    figsize=(10, 6)
)

for systeme in [
    "Collectif",
    "Io",
    "Europa",
    "Ganymede"
]:

    sous_table = resume_multi_pas_detaille[
        resume_multi_pas_detaille["Systeme"]
        ==
        systeme
    ]

    plt.plot(
        sous_table["Delta_t_heures"],
        sous_table["S_T_norm_percent"],
        marker="o",
        label=systeme
    )

plt.xlabel(
    "Pas temporel Δt (heures)"
)

plt.ylabel(
    "S_T_norm (%)"
)

plt.title(
    "Robustesse de S_T_norm selon le système"
)

plt.grid(
    alpha=0.3
)

plt.legend()

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 12. Contrôles finaux
# ------------------------------------------------------------

print(
    "\nContrôles finaux :"
)

print(
    "- Tous les pas ont été calculés :",
    len(resultats_multi_pas)
    ==
    len(pas_heures)
)

print(
    "- Toutes les décompositions sont validées :",
    all(
        resultat["decomposition_validee"]
        for resultat in resultats_multi_pas.values()
    )
)

print(
    "- Toutes les moyennes D_T sont positives :",
    np.all(
        resume_multi_pas_collectif["D_T_mean"]
        >
        0
    )
)

print(
    "- Tous les S_T sont non négatifs :",
    np.all(
        resume_multi_pas_collectif["S_T"]
        >=
        0
    )
)

print(
    "- Tous les S_T_norm sont finis :",
    np.isfinite(
        resume_multi_pas_collectif["S_T_norm"]
    ).all()
)
```

### Robustesse aux opérateurs de normalisation

L'observable de transition dépend de l'opérateur de normalisation
appliqué à l'espace d'état. Pour un opérateur \(\mathcal N\), on
définit

\[
\widetilde X_k^{(\mathcal N)}
=
\mathcal N(X_k),
\]

puis

\[
\boxed{
D_T^{(\mathcal N,\Delta t)}(k)
=
\left\|
\widetilde X_{k+\Delta k}^{(\mathcal N)}
-
\widetilde X_k^{(\mathcal N)}
\right\|_2
}
\]

où \(\Delta k\) représente le nombre de pas élémentaires séparant les
deux états.

Trois familles de normalisation sont comparées.

#### Normalisation z-score

Pour chaque composante \(j\),

\[
\mathcal N_{\mathrm{zscore}}(X_{k,j})
=
\frac{X_{k,j}-\mu_j}{\sigma_j}.
\]

Cette normalisation attribue une variance unitaire à chaque composante.

#### Normalisation min-max

Pour chaque composante \(j\),

\[
\mathcal N_{\mathrm{minmax}}(X_{k,j})
=
\frac{
X_{k,j}-X_{j,\min}
}{
X_{j,\max}-X_{j,\min}
}.
\]

Chaque composante est ainsi projetée dans l'intervalle \([0,1]\).

#### Normalisation physique

Pour chaque satellite \(a\), on définit une longueur caractéristique
\(L_a\) et une vitesse caractéristique \(V_a\), puis

\[
\mathcal N_{\mathrm{physique}}
(\mathbf r_a,\mathbf v_a)
=
\left(
\frac{\mathbf r_a}{L_a},
\frac{\mathbf v_a}{V_a}
\right).
\]

Dans cette première expérience, les échelles caractéristiques sont
estimées à partir de la médiane du rayon orbital et de la médiane de la
norme de vitesse sur la fenêtre étudiée :

\[
L_a
=
\operatorname{median}
\left(
\|\mathbf r_a(k)\|
\right),
\]

\[
V_a
=
\operatorname{median}
\left(
\|\mathbf v_a(k)\|
\right).
\]

Cette normalisation est physiquement interprétable, mais reste
dépendante de la fenêtre de données. Elle constitue donc une première
normalisation physique de benchmark, et non encore une normalisation
canonique définitive de GVH.

```python
# ============================================================
# Robustesse de D_T, S_T et S_T_norm aux normalisations
#
# Normalisations comparées :
# 1. z-score
# 2. min-max
# 3. physique
#
# Expérience factorielle :
# normalisation × pas temporel
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ------------------------------------------------------------
# 1. Préparation et contrôles de l'état collectif brut
# ------------------------------------------------------------

X_brut = np.asarray(
    X_collectif,
    dtype=float
)

if X_brut.ndim != 2:
    raise ValueError(
        "X_collectif doit être une matrice bidimensionnelle."
    )

if X_brut.shape[1] != 18:
    raise ValueError(
        "X_collectif doit contenir exactement 18 composantes."
    )

if not np.isfinite(X_brut).all():
    raise ValueError(
        "X_collectif contient des valeurs non finies."
    )

n_etats, dimension = X_brut.shape

print("Nombre d'états bruts :", n_etats)
print("Dimension de l'état collectif :", dimension)


# ------------------------------------------------------------
# 2. Normalisation z-score
# ------------------------------------------------------------

def normalisation_zscore(X):
    """
    Normalisation composante par composante :

        X_norm = (X - moyenne) / écart-type

    L'écart-type population est utilisé : ddof = 0.
    """

    moyenne = np.mean(
        X,
        axis=0
    )

    ecart_type = np.std(
        X,
        axis=0,
        ddof=0
    )

    if np.any(
        np.isclose(ecart_type, 0.0)
    ):
        indices = np.where(
            np.isclose(ecart_type, 0.0)
        )[0]

        raise ValueError(
            "Écart-type nul pour les composantes : "
            f"{indices.tolist()}"
        )

    X_norm = (
        X - moyenne
    ) / ecart_type

    parametres = {
        "moyenne": moyenne,
        "ecart_type": ecart_type
    }

    return X_norm, parametres


# ------------------------------------------------------------
# 3. Normalisation min-max
# ------------------------------------------------------------

def normalisation_minmax(X):
    """
    Normalisation composante par composante dans [0,1] :

        X_norm = (X - minimum) / (maximum - minimum)
    """

    minimum = np.min(
        X,
        axis=0
    )

    maximum = np.max(
        X,
        axis=0
    )

    amplitude = (
        maximum - minimum
    )

    if np.any(
        np.isclose(amplitude, 0.0)
    ):
        indices = np.where(
            np.isclose(amplitude, 0.0)
        )[0]

        raise ValueError(
            "Amplitude nulle pour les composantes : "
            f"{indices.tolist()}"
        )

    X_norm = (
        X - minimum
    ) / amplitude

    parametres = {
        "minimum": minimum,
        "maximum": maximum,
        "amplitude": amplitude
    }

    return X_norm, parametres


# ------------------------------------------------------------
# 4. Estimation des échelles physiques
# ------------------------------------------------------------

def estimer_echelles_physiques(X):
    """
    Estime, pour chaque satellite :

        L_a = médiane(||r_a||)
        V_a = médiane(||v_a||)

    L'ordre supposé des colonnes est :

        Io        : x, y, z, vx, vy, vz
        Europa    : x, y, z, vx, vy, vz
        Ganymede  : x, y, z, vx, vy, vz
    """

    noms = [
        "Io",
        "Europa",
        "Ganymede"
    ]

    echelles = {}

    for i, nom in enumerate(noms):

        debut = 6 * i

        positions = X[
            :,
            debut:debut + 3
        ]

        vitesses = X[
            :,
            debut + 3:debut + 6
        ]

        rayons = np.linalg.norm(
            positions,
            axis=1
        )

        normes_vitesse = np.linalg.norm(
            vitesses,
            axis=1
        )

        L_a = np.median(
            rayons
        )

        V_a = np.median(
            normes_vitesse
        )

        if (
            not np.isfinite(L_a)
            or
            not np.isfinite(V_a)
            or
            L_a <= 0
            or
            V_a <= 0
        ):
            raise ValueError(
                f"Échelles physiques invalides pour {nom}."
            )

        echelles[nom] = {
            "L": L_a,
            "V": V_a
        }

    return echelles


# ------------------------------------------------------------
# 5. Normalisation physique
# ------------------------------------------------------------

def normalisation_physique(
    X,
    echelles=None
):
    """
    Normalisation par satellite :

        r_tilde = r / L_a
        v_tilde = v / V_a

    Si aucune échelle n'est fournie, elles sont estimées à partir
    des médianes du rayon et de la norme de vitesse.
    """

    if echelles is None:
        echelles = estimer_echelles_physiques(
            X
        )

    noms = [
        "Io",
        "Europa",
        "Ganymede"
    ]

    X_norm = np.empty_like(
        X,
        dtype=float
    )

    for i, nom in enumerate(noms):

        debut = 6 * i

        L_a = echelles[nom]["L"]
        V_a = echelles[nom]["V"]

        X_norm[
            :,
            debut:debut + 3
        ] = (
            X[
                :,
                debut:debut + 3
            ]
            /
            L_a
        )

        X_norm[
            :,
            debut + 3:debut + 6
        ] = (
            X[
                :,
                debut + 3:debut + 6
            ]
            /
            V_a
        )

    return X_norm, echelles


# ------------------------------------------------------------
# 6. Application des trois normalisations
# ------------------------------------------------------------

X_zscore, parametres_zscore = (
    normalisation_zscore(
        X_brut
    )
)

X_minmax, parametres_minmax = (
    normalisation_minmax(
        X_brut
    )
)

echelles_physiques = (
    estimer_echelles_physiques(
        X_brut
    )
)

X_physique, parametres_physiques = (
    normalisation_physique(
        X_brut,
        echelles=echelles_physiques
    )
)

normalisations = {
    "zscore": X_zscore,
    "minmax": X_minmax,
    "physique": X_physique
}


# ------------------------------------------------------------
# 7. Contrôle des normalisations
# ------------------------------------------------------------

print("\nContrôles des normalisations :")

for nom, X_norm in normalisations.items():

    print(
        f"- {nom:10s} | "
        f"forme = {X_norm.shape} | "
        f"finie = {np.isfinite(X_norm).all()}"
    )


# Vérification spécifique du z-score
print(
    "\nZ-score — maximum des moyennes absolues :",
    f"{np.max(np.abs(np.mean(X_zscore, axis=0))):.3e}"
)

print(
    "Z-score — erreur maximale sur les écarts-types :",
    f"{np.max(np.abs(np.std(X_zscore, axis=0) - 1.0)):.3e}"
)


# Vérification spécifique du min-max
print(
    "\nMin-max — minimum global :",
    np.min(X_minmax)
)

print(
    "Min-max — maximum global :",
    np.max(X_minmax)
)


# Tableau des échelles physiques
table_echelles_physiques = pd.DataFrame([
    {
        "Satellite": satellite,
        "L_caracteristique": valeurs["L"],
        "V_caracteristique": valeurs["V"]
    }
    for satellite, valeurs
    in echelles_physiques.items()
])

print(
    "\nÉchelles de la normalisation physique :"
)

display(
    table_echelles_physiques
)


# ------------------------------------------------------------
# 8. Fonction de calcul des observables
# ------------------------------------------------------------

def calculer_observables_normalisation(
    X_norm,
    lag,
    nom_normalisation
):
    """
    Calcule les observables GVH pour une normalisation et un lag.

    Retourne les observables collectives et individuelles.
    """

    if lag <= 0:
        raise ValueError(
            "lag doit être strictement positif."
        )

    if lag >= X_norm.shape[0]:
        raise ValueError(
            "lag est trop grand."
        )

    Delta = (
        X_norm[lag:, :]
        -
        X_norm[:-lag, :]
    )

    groupes = {
        "Collectif": Delta,
        "Io": Delta[:, 0:6],
        "Europa": Delta[:, 6:12],
        "Ganymede": Delta[:, 12:18]
    }

    resultats = {}

    for systeme, Delta_systeme in groupes.items():

        D = np.linalg.norm(
            Delta_systeme,
            axis=1
        )

        moyenne = np.mean(D)

        S_T = np.std(
            D,
            ddof=0
        )

        S_T_norm = (
            S_T / moyenne
            if not np.isclose(moyenne, 0.0)
            else np.nan
        )

        resultats[systeme] = {
            "Normalisation": nom_normalisation,
            "Systeme": systeme,
            "Delta_t_heures": float(lag),
            "N_transitions": D.size,
            "D_T_mean": moyenne,
            "S_T": S_T,
            "S_T_norm": S_T_norm,
            "S_T_norm_percent": 100.0 * S_T_norm,
            "D_T_min": np.min(D),
            "D_T_median": np.median(D),
            "D_T_max": np.max(D),
            "D_T_serie": D
        }

    # Vérification de la décomposition collective
    D_collectif = resultats["Collectif"]["D_T_serie"]
    D_Io = resultats["Io"]["D_T_serie"]
    D_Europa = resultats["Europa"]["D_T_serie"]
    D_Ganymede = resultats["Ganymede"]["D_T_serie"]

    D_reconstruit = np.sqrt(
        D_Io**2
        +
        D_Europa**2
        +
        D_Ganymede**2
    )

    erreur_decomposition = np.max(
        np.abs(
            D_collectif
            -
            D_reconstruit
        )
    )

    decomposition_validee = np.allclose(
        D_collectif,
        D_reconstruit,
        rtol=1e-13,
        atol=1e-13
    )

    return (
        resultats,
        decomposition_validee,
        erreur_decomposition
    )


# ------------------------------------------------------------
# 9. Première comparaison : pas fixe de 1 heure
# ------------------------------------------------------------

lignes_1h = []
series_1h = {}

for nom_normalisation, X_norm in normalisations.items():

    (
        resultats,
        decomposition_validee,
        erreur_decomposition
    ) = calculer_observables_normalisation(
        X_norm=X_norm,
        lag=1,
        nom_normalisation=nom_normalisation
    )

    series_1h[nom_normalisation] = {
        systeme: valeurs["D_T_serie"]
        for systeme, valeurs
        in resultats.items()
    }

    for systeme, valeurs in resultats.items():

        lignes_1h.append({
            "Normalisation":
                nom_normalisation,
            "Systeme":
                systeme,
            "Delta_t_heures":
                valeurs["Delta_t_heures"],
            "N_transitions":
                valeurs["N_transitions"],
            "D_T_mean":
                valeurs["D_T_mean"],
            "S_T":
                valeurs["S_T"],
            "S_T_norm":
                valeurs["S_T_norm"],
            "S_T_norm_percent":
                valeurs["S_T_norm_percent"],
            "D_T_min":
                valeurs["D_T_min"],
            "D_T_median":
                valeurs["D_T_median"],
            "D_T_max":
                valeurs["D_T_max"],
            "Decomposition_validee":
                decomposition_validee,
            "Erreur_decomposition":
                erreur_decomposition
        })

resume_normalisations_1h = pd.DataFrame(
    lignes_1h
)

print(
    "\nComparaison des normalisations à Δt = 1 heure :"
)

display(
    resume_normalisations_1h
)


# ------------------------------------------------------------
# 10. Vérification du z-score historique
# ------------------------------------------------------------

D_zscore_1h = (
    series_1h["zscore"]["Collectif"]
)

verification_zscore_historique = np.allclose(
    D_zscore_1h,
    D_T,
    rtol=1e-14,
    atol=1e-14
)

erreur_zscore_historique = np.max(
    np.abs(
        D_zscore_1h
        -
        D_T
    )
)

print(
    "\nVérification du benchmark historique :"
)

print(
    "Le z-score reproduit exactement D_T initial :",
    verification_zscore_historique
)

print(
    "Erreur maximale :",
    f"{erreur_zscore_historique:.3e}"
)


# ------------------------------------------------------------
# 11. Corrélations temporelles entre normalisations
# ------------------------------------------------------------

noms_normalisations = list(
    normalisations.keys()
)

matrice_correlation = np.empty(
    (
        len(noms_normalisations),
        len(noms_normalisations)
    )
)

for i, nom_a in enumerate(
    noms_normalisations
):

    for j, nom_b in enumerate(
        noms_normalisations
    ):

        D_a = series_1h[
            nom_a
        ]["Collectif"]

        D_b = series_1h[
            nom_b
        ]["Collectif"]

        matrice_correlation[
            i,
            j
        ] = np.corrcoef(
            D_a,
            D_b
        )[0, 1]

correlations_normalisations = pd.DataFrame(
    matrice_correlation,
    index=noms_normalisations,
    columns=noms_normalisations
)

print(
    "\nCorrélations de D_T collectif entre normalisations :"
)

display(
    correlations_normalisations
)


# ------------------------------------------------------------
# 12. Comparaison des rangs temporels
# ------------------------------------------------------------

def correlation_spearman_sans_scipy(
    x,
    y
):
    """
    Corrélation de Spearman calculée à partir des rangs pandas.
    """

    rang_x = pd.Series(
        x
    ).rank(
        method="average"
    )

    rang_y = pd.Series(
        y
    ).rank(
        method="average"
    )

    return np.corrcoef(
        rang_x,
        rang_y
    )[0, 1]


lignes_spearman = []

for i, nom_a in enumerate(
    noms_normalisations
):

    for nom_b in noms_normalisations[
        i + 1:
    ]:

        rho = correlation_spearman_sans_scipy(
            series_1h[nom_a]["Collectif"],
            series_1h[nom_b]["Collectif"]
        )

        lignes_spearman.append({
            "Normalisation_A": nom_a,
            "Normalisation_B": nom_b,
            "Spearman_rho": rho
        })

correlations_spearman = pd.DataFrame(
    lignes_spearman
)

print(
    "\nConservation du classement temporel des transitions :"
)

display(
    correlations_spearman
)


# ------------------------------------------------------------
# 13. Expérience complète :
#     normalisation × pas temporel
# ------------------------------------------------------------

pas_heures = [
    1,
    2,
    3,
    6,
    12,
    24
]

lignes_factorielles = []

for nom_normalisation, X_norm in normalisations.items():

    for lag in pas_heures:

        (
            resultats,
            decomposition_validee,
            erreur_decomposition
        ) = calculer_observables_normalisation(
            X_norm=X_norm,
            lag=lag,
            nom_normalisation=nom_normalisation
        )

        for systeme, valeurs in resultats.items():

            lignes_factorielles.append({
                "Normalisation":
                    nom_normalisation,
                "Systeme":
                    systeme,
                "Delta_t_heures":
                    float(lag),
                "N_transitions":
                    valeurs["N_transitions"],
                "D_T_mean":
                    valeurs["D_T_mean"],
                "S_T":
                    valeurs["S_T"],
                "S_T_norm":
                    valeurs["S_T_norm"],
                "S_T_norm_percent":
                    valeurs["S_T_norm_percent"],
                "D_T_mean_par_heure":
                    valeurs["D_T_mean"]
                    /
                    lag,
                "S_T_par_heure":
                    valeurs["S_T"]
                    /
                    lag,
                "Decomposition_validee":
                    decomposition_validee,
                "Erreur_decomposition":
                    erreur_decomposition
            })

resume_normalisation_temps = pd.DataFrame(
    lignes_factorielles
)

print(
    "\nExpérience factorielle normalisation × pas temporel :"
)

display(
    resume_normalisation_temps
)


# ------------------------------------------------------------
# 14. Tableau collectif uniquement
# ------------------------------------------------------------

resume_collectif_normalisation_temps = (
    resume_normalisation_temps[
        resume_normalisation_temps["Systeme"]
        ==
        "Collectif"
    ]
    .reset_index(
        drop=True
    )
)

print(
    "\nRésultats collectifs :"
)

display(
    resume_collectif_normalisation_temps
)


# ------------------------------------------------------------
# 15. Ratios par rapport au z-score
# ------------------------------------------------------------

reference_zscore = (
    resume_collectif_normalisation_temps[
        resume_collectif_normalisation_temps[
            "Normalisation"
        ]
        ==
        "zscore"
    ][
        [
            "Delta_t_heures",
            "D_T_mean",
            "S_T",
            "S_T_norm"
        ]
    ]
    .rename(
        columns={
            "D_T_mean":
                "D_T_mean_zscore",
            "S_T":
                "S_T_zscore",
            "S_T_norm":
                "S_T_norm_zscore"
        }
    )
)

comparaison_reference = (
    resume_collectif_normalisation_temps
    .merge(
        reference_zscore,
        on="Delta_t_heures",
        how="left"
    )
)

comparaison_reference[
    "Ratio_D_T_vs_zscore"
] = (
    comparaison_reference["D_T_mean"]
    /
    comparaison_reference["D_T_mean_zscore"]
)

comparaison_reference[
    "Ratio_S_T_vs_zscore"
] = (
    comparaison_reference["S_T"]
    /
    comparaison_reference["S_T_zscore"]
)

comparaison_reference[
    "Ratio_S_T_norm_vs_zscore"
] = (
    comparaison_reference["S_T_norm"]
    /
    comparaison_reference["S_T_norm_zscore"]
)

print(
    "\nComparaison avec la référence z-score :"
)

display(
    comparaison_reference[
        [
            "Normalisation",
            "Delta_t_heures",
            "D_T_mean",
            "S_T",
            "S_T_norm",
            "Ratio_D_T_vs_zscore",
            "Ratio_S_T_vs_zscore",
            "Ratio_S_T_norm_vs_zscore"
        ]
    ]
)


# ------------------------------------------------------------
# 16. Graphique :
#     D_T moyen collectif selon la normalisation
# ------------------------------------------------------------

plt.figure(
    figsize=(9, 5)
)

for nom_normalisation in noms_normalisations:

    sous_table = (
        resume_collectif_normalisation_temps[
            resume_collectif_normalisation_temps[
                "Normalisation"
            ]
            ==
            nom_normalisation
        ]
    )

    plt.plot(
        sous_table["Delta_t_heures"],
        sous_table["D_T_mean"],
        marker="o",
        label=nom_normalisation
    )

plt.xlabel(
    "Pas temporel Δt (heures)"
)

plt.ylabel(
    "D_T moyen collectif"
)

plt.title(
    "Effet conjoint de la normalisation et du pas temporel"
)

plt.grid(
    alpha=0.3
)

plt.legend()

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 17. Graphique :
#     S_T_norm collectif selon la normalisation
# ------------------------------------------------------------

plt.figure(
    figsize=(9, 5)
)

for nom_normalisation in noms_normalisations:

    sous_table = (
        resume_collectif_normalisation_temps[
            resume_collectif_normalisation_temps[
                "Normalisation"
            ]
            ==
            nom_normalisation
        ]
    )

    plt.plot(
        sous_table["Delta_t_heures"],
        sous_table["S_T_norm_percent"],
        marker="o",
        label=nom_normalisation
    )

plt.xlabel(
    "Pas temporel Δt (heures)"
)

plt.ylabel(
    "S_T_norm collectif (%)"
)

plt.title(
    "Robustesse de la variabilité relative aux normalisations"
)

plt.grid(
    alpha=0.3
)

plt.legend()

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 18. Contrôles finaux
# ------------------------------------------------------------

print(
    "\nContrôles finaux :"
)

print(
    "- Toutes les normalisations sont finies :",
    all(
        np.isfinite(X_norm).all()
        for X_norm in normalisations.values()
    )
)

print(
    "- Toutes les décompositions sont validées :",
    resume_normalisation_temps[
        "Decomposition_validee"
    ].all()
)

print(
    "- Toutes les moyennes D_T sont positives :",
    (
        resume_normalisation_temps[
            "D_T_mean"
        ]
        >
        0
    ).all()
)

print(
    "- Tous les S_T sont non négatifs :",
    (
        resume_normalisation_temps[
            "S_T"
        ]
        >=
        0
    ).all()
)

print(
    "- Tous les S_T_norm sont finis :",
    np.isfinite(
        resume_normalisation_temps[
            "S_T_norm"
        ]
    ).all()
)
```

### Conclusion du test de robustesse aux normalisations

Les observables GVH ont été calculées avec trois opérateurs de
normalisation :

\[
\mathcal N_{\mathrm{zscore}},
\qquad
\mathcal N_{\mathrm{minmax}},
\qquad
\mathcal N_{\mathrm{physique}}.
\]

Les valeurs absolues de \(D_T\) et de \(S_T\) changent sensiblement
selon la normalisation. À une heure, la moyenne collective vaut

\[
0.414467
\]

pour le z-score,

\[
0.145656
\]

pour le min-max, et

\[
0.239295
\]

pour la normalisation physique.

Ces différences confirment que \(D_T\) et \(S_T\) dépendent de la
géométrie métrique imposée par l'opérateur de normalisation.

En revanche, la structure temporelle des transitions est fortement
conservée. Les corrélations linéaires entre les séries collectives sont
supérieures à \(0.993\), et les corrélations de rang sont supérieures à
\(0.988\).

La variabilité relative

\[
S_T^{\mathrm{norm}}
=
\frac{S_T}{\overline{D_T}}
\]

présente une robustesse encore plus forte. À une heure, elle vaut

\[
0.389119\%
\]

pour le z-score,

\[
0.389229\%
\]

pour le min-max, et

\[
0.385660\%
\]

pour la normalisation physique.

Les trois opérateurs conservent également la hiérarchie individuelle

\[
S_{T,\mathrm{Europa}}^{\mathrm{norm}}
>
S_{T,\mathrm{Io}}^{\mathrm{norm}}
>
S_{T,\mathrm{Ganymède}}^{\mathrm{norm}}
\]

au pas d'une heure.

L'expérience factorielle montre enfin que la diminution de
\(S_T^{\mathrm{norm}}\) lorsque le pas temporel augmente est commune
aux trois normalisations. L'écart entre les géométries devient toutefois
plus visible à vingt-quatre heures, particulièrement pour la
normalisation physique.

Ces résultats permettent de distinguer deux niveaux :

\[
\text{les amplitudes absolues, dépendantes de la normalisation,}
\]

et

\[
\text{la structure relative des transitions, largement robuste.}
\]

Le résultat principal de cette expérience est donc que
\(S_T^{\mathrm{norm}}\), le classement temporel de \(D_T\) et les
tendances multi-échelles sont nettement plus robustes que les valeurs
absolues de \(D_T\) et de \(S_T\).

### Validation hors échantillon des normalisations

Les états sont séparés chronologiquement en deux ensembles :

\[
\mathcal D_{\mathrm{train}}
=
\{X_0,\ldots,X_{n_{\mathrm{train}}-1}\},
\]

et

\[
\mathcal D_{\mathrm{test}}
=
\{X_{n_{\mathrm{train}}},\ldots,X_{N-1}\}.
\]

Pour chaque opérateur de normalisation, les paramètres sont estimés
exclusivement sur la période d'apprentissage :

\[
\widehat\Theta_{\mathcal N}
=
\operatorname{fit}
\left(
\mathcal D_{\mathrm{train}}
\right).
\]

Ces paramètres sont ensuite gelés et appliqués aux deux périodes :

\[
\widetilde X_k^{\mathrm{train}}
=
\mathcal N_{\widehat\Theta_{\mathcal N}}
\left(
X_k^{\mathrm{train}}
\right),
\]

\[
\widetilde X_k^{\mathrm{test}}
=
\mathcal N_{\widehat\Theta_{\mathcal N}}
\left(
X_k^{\mathrm{test}}
\right).
\]

Aucun paramètre de normalisation n'est recalculé sur la période de
test.

Pour chaque période, on calcule

\[
D_T(k)
=
\left\|
\widetilde X_{k+1}
-
\widetilde X_k
\right\|_2,
\]

\[
S_T
=
\operatorname{std}(D_T),
\]

et

\[
S_T^{\mathrm{norm}}
=
\frac{S_T}{\overline{D_T}}.
\]

La stabilité hors échantillon est évaluée à partir des rapports

\[
R_D
=
\frac{
\overline{D_T}^{\,\mathrm{test}}
}{
\overline{D_T}^{\,\mathrm{train}}
},
\]

\[
R_S
=
\frac{
S_T^{\mathrm{test}}
}{
S_T^{\mathrm{train}}
},
\]

et

\[
R_{S,\mathrm{norm}}
=
\frac{
S_{T,\mathrm{norm}}^{\mathrm{test}}
}{
S_{T,\mathrm{norm}}^{\mathrm{train}}
}.
\]

Des rapports proches de \(1\) indiquent une stabilité descriptive entre
les deux périodes. Ils ne constituent cependant pas, à eux seuls, une
preuve d'invariance physique.

```python
# ============================================================
# Validation hors échantillon des normalisations GVH
#
# Séparation chronologique :
# - 80 % apprentissage
# - 20 % test
#
# Les paramètres sont calculés sur train uniquement,
# puis appliqués sans recalcul sur test.
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ------------------------------------------------------------
# 1. Préparation des données
# ------------------------------------------------------------

X_brut = np.asarray(
    X_collectif,
    dtype=float
)

if X_brut.ndim != 2:
    raise ValueError(
        "X_collectif doit être une matrice bidimensionnelle."
    )

if X_brut.shape[1] != 18:
    raise ValueError(
        "X_collectif doit contenir 18 composantes."
    )

if not np.isfinite(X_brut).all():
    raise ValueError(
        "X_collectif contient des valeurs non finies."
    )

N_etats = X_brut.shape[0]

fraction_train = 0.80

indice_separation = int(
    np.floor(
        fraction_train * N_etats
    )
)

if indice_separation < 2:
    raise ValueError(
        "La période d'apprentissage est trop courte."
    )

if N_etats - indice_separation < 2:
    raise ValueError(
        "La période de test est trop courte."
    )

X_train_brut = X_brut[
    :indice_separation
]

X_test_brut = X_brut[
    indice_separation:
]

print("Séparation chronologique :")
print(
    f"- Nombre total d'états : {N_etats}"
)
print(
    f"- États d'apprentissage : {X_train_brut.shape[0]}"
)
print(
    f"- États de test : {X_test_brut.shape[0]}"
)
print(
    f"- Indice de séparation : {indice_separation}"
)


# ------------------------------------------------------------
# 2. Ajustement et application du z-score
# ------------------------------------------------------------

def ajuster_zscore(X_train):
    moyenne = np.mean(
        X_train,
        axis=0
    )

    ecart_type = np.std(
        X_train,
        axis=0,
        ddof=0
    )

    if np.any(
        np.isclose(ecart_type, 0.0)
    ):
        indices = np.where(
            np.isclose(ecart_type, 0.0)
        )[0]

        raise ValueError(
            "Écart-type nul dans train pour les composantes : "
            f"{indices.tolist()}"
        )

    return {
        "moyenne": moyenne,
        "ecart_type": ecart_type
    }


def appliquer_zscore(X, parametres):
    return (
        X - parametres["moyenne"]
    ) / parametres["ecart_type"]


# ------------------------------------------------------------
# 3. Ajustement et application du min-max
# ------------------------------------------------------------

def ajuster_minmax(X_train):
    minimum = np.min(
        X_train,
        axis=0
    )

    maximum = np.max(
        X_train,
        axis=0
    )

    amplitude = (
        maximum - minimum
    )

    if np.any(
        np.isclose(amplitude, 0.0)
    ):
        indices = np.where(
            np.isclose(amplitude, 0.0)
        )[0]

        raise ValueError(
            "Amplitude nulle dans train pour les composantes : "
            f"{indices.tolist()}"
        )

    return {
        "minimum": minimum,
        "maximum": maximum,
        "amplitude": amplitude
    }


def appliquer_minmax(X, parametres):
    return (
        X - parametres["minimum"]
    ) / parametres["amplitude"]


# ------------------------------------------------------------
# 4. Ajustement et application de la normalisation physique
# ------------------------------------------------------------

def ajuster_normalisation_physique(X_train):
    noms = [
        "Io",
        "Europa",
        "Ganymede"
    ]

    echelles = {}

    for i, nom in enumerate(noms):

        debut = 6 * i

        positions = X_train[
            :,
            debut:debut + 3
        ]

        vitesses = X_train[
            :,
            debut + 3:debut + 6
        ]

        rayons = np.linalg.norm(
            positions,
            axis=1
        )

        normes_vitesse = np.linalg.norm(
            vitesses,
            axis=1
        )

        L_a = np.median(
            rayons
        )

        V_a = np.median(
            normes_vitesse
        )

        if (
            not np.isfinite(L_a)
            or
            not np.isfinite(V_a)
            or
            L_a <= 0
            or
            V_a <= 0
        ):
            raise ValueError(
                f"Échelles physiques invalides pour {nom}."
            )

        echelles[nom] = {
            "L": L_a,
            "V": V_a
        }

    return echelles


def appliquer_normalisation_physique(
    X,
    echelles
):
    noms = [
        "Io",
        "Europa",
        "Ganymede"
    ]

    X_norm = np.empty_like(
        X,
        dtype=float
    )

    for i, nom in enumerate(noms):

        debut = 6 * i

        L_a = echelles[nom]["L"]
        V_a = echelles[nom]["V"]

        X_norm[
            :,
            debut:debut + 3
        ] = (
            X[
                :,
                debut:debut + 3
            ]
            /
            L_a
        )

        X_norm[
            :,
            debut + 3:debut + 6
        ] = (
            X[
                :,
                debut + 3:debut + 6
            ]
            /
            V_a
        )

    return X_norm


# ------------------------------------------------------------
# 5. Ajustement sur train uniquement
# ------------------------------------------------------------

parametres_zscore_train = ajuster_zscore(
    X_train_brut
)

parametres_minmax_train = ajuster_minmax(
    X_train_brut
)

parametres_physiques_train = (
    ajuster_normalisation_physique(
        X_train_brut
    )
)


# ------------------------------------------------------------
# 6. Application des paramètres gelés
# ------------------------------------------------------------

X_train_zscore = appliquer_zscore(
    X_train_brut,
    parametres_zscore_train
)

X_test_zscore = appliquer_zscore(
    X_test_brut,
    parametres_zscore_train
)

X_train_minmax = appliquer_minmax(
    X_train_brut,
    parametres_minmax_train
)

X_test_minmax = appliquer_minmax(
    X_test_brut,
    parametres_minmax_train
)

X_train_physique = appliquer_normalisation_physique(
    X_train_brut,
    parametres_physiques_train
)

X_test_physique = appliquer_normalisation_physique(
    X_test_brut,
    parametres_physiques_train
)

donnees_normalisees = {
    "zscore": {
        "train": X_train_zscore,
        "test": X_test_zscore
    },
    "minmax": {
        "train": X_train_minmax,
        "test": X_test_minmax
    },
    "physique": {
        "train": X_train_physique,
        "test": X_test_physique
    }
}


# ------------------------------------------------------------
# 7. Vérification de l'absence de recalcul sur test
# ------------------------------------------------------------

print(
    "\nContrôles des ensembles normalisés :"
)

for normalisation, periodes in donnees_normalisees.items():

    for periode, X_norm in periodes.items():

        print(
            f"- {normalisation:9s} | "
            f"{periode:5s} | "
            f"forme = {X_norm.shape} | "
            f"finie = {np.isfinite(X_norm).all()}"
        )


# Un test min-max hors échantillon peut légitimement sortir de [0,1].
test_minmax_hors_intervalle = np.mean(
    (X_test_minmax < 0.0)
    |
    (X_test_minmax > 1.0)
)

print(
    "\nProportion des valeurs min-max de test hors [0,1] :",
    f"{100 * test_minmax_hors_intervalle:.6f} %"
)


# ------------------------------------------------------------
# 8. Fonction de calcul des observables
# ------------------------------------------------------------

def calculer_observables_periode(
    X_norm,
    normalisation,
    periode,
    lag=1
):
    """
    Calcule D_T, S_T et S_T_norm pour une période.
    Les transitions restent internes à la période considérée.
    """

    if lag <= 0:
        raise ValueError(
            "lag doit être strictement positif."
        )

    if lag >= X_norm.shape[0]:
        raise ValueError(
            f"{periode} : lag trop grand."
        )

    Delta = (
        X_norm[lag:]
        -
        X_norm[:-lag]
    )

    groupes = {
        "Collectif": Delta,
        "Io": Delta[:, 0:6],
        "Europa": Delta[:, 6:12],
        "Ganymede": Delta[:, 12:18]
    }

    resultats = {}
    series = {}

    for systeme, Delta_systeme in groupes.items():

        D = np.linalg.norm(
            Delta_systeme,
            axis=1
        )

        D_mean = np.mean(D)

        S_T = np.std(
            D,
            ddof=0
        )

        S_T_norm = (
            S_T / D_mean
            if not np.isclose(D_mean, 0.0)
            else np.nan
        )

        resultats[systeme] = {
            "Normalisation": normalisation,
            "Periode": periode,
            "Systeme": systeme,
            "Delta_t_heures": float(lag),
            "N_etats": X_norm.shape[0],
            "N_transitions": D.size,
            "D_T_mean": D_mean,
            "S_T": S_T,
            "S_T_norm": S_T_norm,
            "S_T_norm_percent":
                100.0 * S_T_norm,
            "D_T_min": np.min(D),
            "D_T_median": np.median(D),
            "D_T_max": np.max(D)
        }

        series[systeme] = D

    # Décomposition quadratique
    D_collectif = series["Collectif"]

    D_reconstruit = np.sqrt(
        series["Io"]**2
        +
        series["Europa"]**2
        +
        series["Ganymede"]**2
    )

    erreur_decomposition = np.max(
        np.abs(
            D_collectif
            -
            D_reconstruit
        )
    )

    decomposition_validee = np.allclose(
        D_collectif,
        D_reconstruit,
        rtol=1e-13,
        atol=1e-13
    )

    return (
        resultats,
        series,
        decomposition_validee,
        erreur_decomposition
    )


# ------------------------------------------------------------
# 9. Calcul train/test à Δt = 1 heure
# ------------------------------------------------------------

lignes_resultats = []
series_hors_echantillon = {}

for normalisation, periodes in donnees_normalisees.items():

    series_hors_echantillon[
        normalisation
    ] = {}

    for periode, X_norm in periodes.items():

        (
            resultats,
            series,
            decomposition_validee,
            erreur_decomposition
        ) = calculer_observables_periode(
            X_norm=X_norm,
            normalisation=normalisation,
            periode=periode,
            lag=1
        )

        series_hors_echantillon[
            normalisation
        ][periode] = series

        for systeme, valeurs in resultats.items():

            ligne = valeurs.copy()

            ligne[
                "Decomposition_validee"
            ] = decomposition_validee

            ligne[
                "Erreur_decomposition"
            ] = erreur_decomposition

            lignes_resultats.append(
                ligne
            )

resume_hors_echantillon = pd.DataFrame(
    lignes_resultats
)

print(
    "\nObservables hors échantillon :"
)

display(
    resume_hors_echantillon
)


# ------------------------------------------------------------
# 10. Tableau collectif train/test
# ------------------------------------------------------------

resume_collectif_hors_echantillon = (
    resume_hors_echantillon[
        resume_hors_echantillon["Systeme"]
        ==
        "Collectif"
    ]
    .reset_index(
        drop=True
    )
)

print(
    "\nRésultats collectifs train/test :"
)

display(
    resume_collectif_hors_echantillon
)


# ------------------------------------------------------------
# 11. Comparaison test / train
# ------------------------------------------------------------

table_train = (
    resume_hors_echantillon[
        resume_hors_echantillon["Periode"]
        ==
        "train"
    ][
        [
            "Normalisation",
            "Systeme",
            "D_T_mean",
            "S_T",
            "S_T_norm"
        ]
    ]
    .rename(
        columns={
            "D_T_mean": "D_T_mean_train",
            "S_T": "S_T_train",
            "S_T_norm": "S_T_norm_train"
        }
    )
)

table_test = (
    resume_hors_echantillon[
        resume_hors_echantillon["Periode"]
        ==
        "test"
    ][
        [
            "Normalisation",
            "Systeme",
            "D_T_mean",
            "S_T",
            "S_T_norm"
        ]
    ]
    .rename(
        columns={
            "D_T_mean": "D_T_mean_test",
            "S_T": "S_T_test",
            "S_T_norm": "S_T_norm_test"
        }
    )
)

comparaison_train_test = table_train.merge(
    table_test,
    on=[
        "Normalisation",
        "Systeme"
    ],
    how="inner"
)

comparaison_train_test[
    "Ratio_D_T_test_train"
] = (
    comparaison_train_test["D_T_mean_test"]
    /
    comparaison_train_test["D_T_mean_train"]
)

comparaison_train_test[
    "Ratio_S_T_test_train"
] = (
    comparaison_train_test["S_T_test"]
    /
    comparaison_train_test["S_T_train"]
)

comparaison_train_test[
    "Ratio_S_T_norm_test_train"
] = (
    comparaison_train_test["S_T_norm_test"]
    /
    comparaison_train_test["S_T_norm_train"]
)

comparaison_train_test[
    "Variation_D_T_percent"
] = (
    100.0
    *
    (
        comparaison_train_test[
            "Ratio_D_T_test_train"
        ]
        -
        1.0
    )
)

comparaison_train_test[
    "Variation_S_T_percent"
] = (
    100.0
    *
    (
        comparaison_train_test[
            "Ratio_S_T_test_train"
        ]
        -
        1.0
    )
)

comparaison_train_test[
    "Variation_S_T_norm_percent"
] = (
    100.0
    *
    (
        comparaison_train_test[
            "Ratio_S_T_norm_test_train"
        ]
        -
        1.0
    )
)

print(
    "\nComparaison test / train :"
)

display(
    comparaison_train_test
)


# ------------------------------------------------------------
# 12. Tableau collectif compact
# ------------------------------------------------------------

comparaison_collective = (
    comparaison_train_test[
        comparaison_train_test["Systeme"]
        ==
        "Collectif"
    ][
        [
            "Normalisation",
            "D_T_mean_train",
            "D_T_mean_test",
            "Variation_D_T_percent",
            "S_T_train",
            "S_T_test",
            "Variation_S_T_percent",
            "S_T_norm_train",
            "S_T_norm_test",
            "Variation_S_T_norm_percent"
        ]
    ]
    .reset_index(
        drop=True
    )
)

print(
    "\nSynthèse collective hors échantillon :"
)

display(
    comparaison_collective
)


# ------------------------------------------------------------
# 13. Corrélations entre normalisations sur le test
# ------------------------------------------------------------

noms_normalisations = [
    "zscore",
    "minmax",
    "physique"
]

matrice_corr_test = np.empty(
    (
        len(noms_normalisations),
        len(noms_normalisations)
    )
)

for i, nom_a in enumerate(
    noms_normalisations
):

    for j, nom_b in enumerate(
        noms_normalisations
    ):

        D_a = (
            series_hors_echantillon[
                nom_a
            ]["test"]["Collectif"]
        )

        D_b = (
            series_hors_echantillon[
                nom_b
            ]["test"]["Collectif"]
        )

        matrice_corr_test[
            i,
            j
        ] = np.corrcoef(
            D_a,
            D_b
        )[0, 1]

correlations_test = pd.DataFrame(
    matrice_corr_test,
    index=noms_normalisations,
    columns=noms_normalisations
)

print(
    "\nCorrélations entre normalisations sur le test :"
)

display(
    correlations_test
)


# ------------------------------------------------------------
# 14. Vérification de la hiérarchie des satellites
# ------------------------------------------------------------

hierarchies = []

for normalisation in noms_normalisations:

    for periode in [
        "train",
        "test"
    ]:

        sous_table = resume_hors_echantillon[
            (
                resume_hors_echantillon[
                    "Normalisation"
                ]
                ==
                normalisation
            )
            &
            (
                resume_hors_echantillon[
                    "Periode"
                ]
                ==
                periode
            )
            &
            (
                resume_hors_echantillon[
                    "Systeme"
                ]
                !=
                "Collectif"
            )
        ].sort_values(
            "S_T_norm",
            ascending=False
        )

        ordre = " > ".join(
            sous_table["Systeme"].tolist()
        )

        hierarchies.append({
            "Normalisation": normalisation,
            "Periode": periode,
            "Hierarchie_S_T_norm": ordre
        })

table_hierarchies = pd.DataFrame(
    hierarchies
)

print(
    "\nHiérarchie de S_T_norm :"
)

display(
    table_hierarchies
)


# ------------------------------------------------------------
# 15. Graphique train/test de S_T_norm collectif
# ------------------------------------------------------------

table_graphique = (
    resume_collectif_hors_echantillon
    .pivot(
        index="Normalisation",
        columns="Periode",
        values="S_T_norm_percent"
    )
)

table_graphique.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.xlabel(
    "Normalisation"
)

plt.ylabel(
    "S_T_norm collectif (%)"
)

plt.title(
    "Stabilité hors échantillon de S_T_norm"
)

plt.xticks(
    rotation=0
)

plt.tight_layout()

plt.show()


# ------------------------------------------------------------
# 16. Contrôles finaux
# ------------------------------------------------------------

print(
    "\nContrôles finaux :"
)

print(
    "- Paramètres ajustés uniquement sur train : True"
)

print(
    "- Toutes les matrices sont finies :",
    all(
        np.isfinite(X_norm).all()
        for periodes
        in donnees_normalisees.values()
        for X_norm
        in periodes.values()
    )
)

print(
    "- Toutes les décompositions sont validées :",
    resume_hors_echantillon[
        "Decomposition_validee"
    ].all()
)

print(
    "- Tous les D_T moyens sont positifs :",
    (
        resume_hors_echantillon[
            "D_T_mean"
        ]
        >
        0
    ).all()
)

print(
    "- Tous les S_T sont non négatifs :",
    (
        resume_hors_echantillon[
            "S_T"
        ]
        >=
        0
    ).all()
)

print(
    "- Tous les S_T_norm sont finis :",
    np.isfinite(
        resume_hors_echantillon[
            "S_T_norm"
        ]
    ).all()
)
```

### Conclusion de la validation hors échantillon

Les paramètres des trois opérateurs de normalisation ont été estimés
exclusivement sur les premiers \(80\%\) de la trajectoire, puis
appliqués sans recalcul aux \(20\%\) restants.

Au niveau collectif, la moyenne de la norme de transition présente une
très forte stabilité. La variation entre apprentissage et test vaut

\[
+0.0168\%
\]

pour le z-score,

\[
+0.0204\%
\]

pour le min-max, et

\[
+0.0042\%
\]

pour la normalisation physique.

La variabilité relative collective reste également stable. Sa variation
est de

\[
+1.61\%
\]

pour le z-score,

\[
+1.93\%
\]

pour le min-max, et

\[
+0.89\%
\]

pour la normalisation physique.

Les corrélations entre les séries de transition obtenues sur la période
de test restent supérieures à \(0.992\), ce qui montre que les trois
normalisations conservent presque la même structure temporelle hors
échantillon.

La hiérarchie individuelle

\[
\mathrm{Europa}
>
\mathrm{Io}
>
\mathrm{Ganymède}
\]

en termes de \(S_T^{\mathrm{norm}}\) est conservée sur train et test
pour les trois normalisations.

Les résultats individuels montrent toutefois une hétérogénéité
temporelle plus marquée. Europa présente une diminution modérée de sa
variabilité relative, tandis que Ganymède montre une augmentation
relative d'environ \(34\%\). Cette hausse doit être nuancée par le fait
que la valeur absolue de \(S_T^{\mathrm{norm}}\) de Ganymède reste très
faible.

La validation hors échantillon soutient donc fortement la robustesse
collective de \(D_T\), \(S_T\) et \(S_T^{\mathrm{norm}}\) lorsque la
géométrie de normalisation est fixée à l'avance. Elle ne démontre pas
encore une invariance universelle, mais elle réduit nettement le risque
que les résultats précédents proviennent d'un ajustement rétrospectif
sur l'ensemble complet des données.

```python
import numpy as np
import pandas as pd

satellites = ["Io", "Europa", "Ganymede"]

transitions_2026 = {}

for satellite in satellites:
    colonnes_xyz = [
        f"Delta_{satellite}_X",
        f"Delta_{satellite}_Y",
        f"Delta_{satellite}_Z"
    ]

    temp = Delta_X_collectif_df[
        ["JDTDB", "Calendar_Date_TDB"] + colonnes_xyz
    ].copy()

    temp["D_T"] = np.linalg.norm(
        temp[colonnes_xyz].to_numpy(dtype=float),
        axis=1
    )

    transitions_2026[satellite] = temp

    print(
        satellite,
        "| N =", len(temp),
        "| D_T_mean =", temp["D_T"].mean(),
        "| D_T_std =", temp["D_T"].std(ddof=0),
        "| D_T_max =", temp["D_T"].max()
    )
```

```python
observables_2026 = []

for satellite, temp in transitions_2026.items():
    D_T = temp["D_T"].to_numpy(dtype=float)

    D_T_mean = D_T.mean()
    S_T = D_T.std(ddof=0)
    D_T_max = D_T.max()

    # Convention utilisée dans le pipeline 2000
    u = D_T / D_T_max

    u_c = np.quantile(u, 0.95)
    D_c = u_c * D_T_max
    R_super = np.mean(u > u_c)

    observables_2026.append({
        "Satellite": satellite,
        "Annee": 2026,
        "N_transitions": len(D_T),
        "D_T_mean": D_T_mean,
        "S_T": S_T,
        "S_T_norm": S_T / D_T_max,
        "D_T_min": D_T.min(),
        "D_T_max": D_T_max,
        "D_c": D_c,
        "u_c": u_c,
        "R_super": R_super
    })

observables_gvh_2026_base = pd.DataFrame(observables_2026)

display(
    observables_gvh_2026_base
    .sort_values("Satellite")
    .reset_index(drop=True)
)
```

```python
import numpy as np
import pandas as pd

angles_2026 = {}
resume_angles_2026 = []

for satellite in ["Io", "Europa", "Ganymede"]:

    colonnes_xyz = [
        f"Delta_{satellite}_X",
        f"Delta_{satellite}_Y",
        f"Delta_{satellite}_Z"
    ]

    vecteurs = (
        Delta_X_collectif_df[colonnes_xyz]
        .to_numpy(dtype=float)
    )

    # Paires de vecteurs successifs
    v1 = vecteurs[:-1]
    v2 = vecteurs[1:]

    normes_v1 = np.linalg.norm(v1, axis=1)
    normes_v2 = np.linalg.norm(v2, axis=1)

    denominateur = normes_v1 * normes_v2

    # Évite toute division par zéro
    masque_valide = denominateur > 0

    cos_theta = np.full(len(denominateur), np.nan)

    cos_theta[masque_valide] = (
        np.sum(v1[masque_valide] * v2[masque_valide], axis=1)
        / denominateur[masque_valide]
    )

    # Protection contre les erreurs d'arrondi numérique
    cos_theta = np.clip(cos_theta, -1.0, 1.0)

    theta_deg = np.degrees(np.arccos(cos_theta))

    angles_df = pd.DataFrame({
        "JDTDB": Delta_X_collectif_df["JDTDB"].iloc[1:].to_numpy(),
        "Calendar_Date_TDB":
            Delta_X_collectif_df["Calendar_Date_TDB"].iloc[1:].to_numpy(),
        "theta_deg": theta_deg
    })

    angles_2026[satellite] = angles_df

    theta_valides = theta_deg[np.isfinite(theta_deg)]

    theta_mean = np.mean(theta_valides)
    theta_median = np.median(theta_valides)
    theta_std = np.std(theta_valides, ddof=0)
    theta_max = np.max(theta_valides)

    R170 = np.mean(theta_valides >= 170)
    R175 = np.mean(theta_valides >= 175)
    R180 = np.mean(theta_valides >= 179.999)

    periode_heures = 360.0 / theta_mean
    periode_jours = periode_heures / 24.0

    resume_angles_2026.append({
        "Satellite": satellite,
        "Annee": 2026,
        "N_angles": len(theta_valides),
        "theta_mean": theta_mean,
        "theta_median": theta_median,
        "theta_std": theta_std,
        "theta_max": theta_max,
        "R170": R170,
        "R175": R175,
        "R180": R180,
        "Periode_heures_estimee": periode_heures,
        "Periode_jours_estimee": periode_jours
    })

resume_angles_2026_df = pd.DataFrame(resume_angles_2026)

display(
    resume_angles_2026_df
    .sort_values("Satellite")
    .reset_index(drop=True)
)
```

```python
observables_gvh_2026_complet = (
    observables_gvh_2026_base
    .merge(
        resume_angles_2026_df,
        on=["Satellite", "Annee"],
        how="inner",
        validate="one_to_one"
    )
    .sort_values("Satellite")
    .reset_index(drop=True)
)

colonnes_ordre = [
    "Satellite",
    "Annee",
    "N_transitions",
    "D_T_mean",
    "S_T",
    "S_T_norm",
    "D_T_min",
    "D_T_max",
    "D_c",
    "u_c",
    "R_super",
    "N_angles",
    "theta_mean",
    "theta_median",
    "theta_std",
    "theta_max",
    "R170",
    "R175",
    "R180",
    "Periode_heures_estimee",
    "Periode_jours_estimee"
]

observables_gvh_2026_complet = (
    observables_gvh_2026_complet[colonnes_ordre]
)

display(observables_gvh_2026_complet)
```
