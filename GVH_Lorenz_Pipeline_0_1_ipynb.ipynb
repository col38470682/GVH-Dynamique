GVH Lorenz Pipeline 0.1

1. Objectif

Ce notebook applique le cadre descriptif GVH Dynamic 0.1 au système dynamique de Lorenz.

L’objectif est de construire une chaîne de calcul reproductible permettant d’extraire, à partir d’une trajectoire tridimensionnelle du système de Lorenz, le même ensemble standardisé d’observables géométriques que celui utilisé dans les autres pipelines GVH.

La chaîne méthodologique est :

[
\text{Système de Lorenz}
;\longrightarrow;
\mathbf{X}(t)
;\longrightarrow;
D_T
;\longrightarrow;
S_T
;\longrightarrow;
D_c
;\longrightarrow;
R_{\rm super}
;\longrightarrow;
(A_H,B_H)
;\longrightarrow;
(\theta_{\rm mean},\theta_{\max},R_{180}).
]

Les observables calculées sont :

* (D_T^{mean}) : distance locale moyenne entre états successifs ;
* (S_T) : dispersion des distances locales ;
* (S_T^{norm}) : dispersion normalisée ;
* (D_c) : seuil critique défini à partir de la distribution de (D_T) ;
* (R_{\rm super}) : fraction des états situés au-dessus du seuil critique ;
* (A_H) et (B_H) : paramètres de la décroissance de (R_{\rm super}(D_c)) ;
* (\theta_{\rm mean}) : angle moyen entre déplacements successifs ;
* (\theta_{\max}) : angle maximal observé ;
* (R_{180}) : fraction des configurations proches de l’opposition angulaire.

Cette version 0.1 est strictement descriptive et méthodologique. Elle vise à tester la transférabilité des observables GVH à un système dynamique non linéaire chaotique.

Aucune interprétation physique dans le cadre du modèle Volume Partagé n’est introduite dans ce notebook.

2. Imports Python

Cette section charge les bibliothèques nécessaires pour :

* intégrer numériquement le système de Lorenz ;
* manipuler les trajectoires sous forme de tableaux ;
* calculer les observables GVH ;
* produire les figures ;
* exporter les résultats.

```python
# ============================================================
# GVH_Lorenz_Pipeline_0.1
# Étape 2 — Imports Python
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.integrate import solve_ivp
from scipy.optimize import curve_fit
```

3. Génération de la trajectoire de Lorenz

Cette section génère une trajectoire tridimensionnelle du système de Lorenz.

Le système est défini par :

[
\frac{dx}{dt}=\sigma(y-x)
]

[
\frac{dy}{dt}=x(\rho-z)-y
]

[
\frac{dz}{dt}=xy-\beta z
]

Pour cette version 0.1, on utilise les paramètres classiques :

[
\sigma=10,\quad \rho=28,\quad \beta=\frac{8}{3}.
]

Ces paramètres produisent une dynamique chaotique standard, utilisée ici comme cas de référence pour tester les observables GVH.

```python
# ============================================================
# Étape 3 — Génération de la trajectoire de Lorenz
# ============================================================

# Paramètres classiques du système de Lorenz
sigma = 10.0
rho = 28.0
beta = 8.0 / 3.0

# Définition du système de Lorenz
def lorenz_system(t, state, sigma, rho, beta):
    x, y, z = state

    dxdt = sigma * (y - x)
    dydt = x * (rho - z) - y
    dzdt = x * y - beta * z

    return [dxdt, dydt, dzdt]

# Conditions initiales
initial_state = [1.0, 1.0, 1.0]

# Intervalle temporel
t_start = 0.0
t_end = 50.0
n_points = 10000

t_eval = np.linspace(t_start, t_end, n_points)

# Intégration numérique
solution = solve_ivp(
    fun=lorenz_system,
    t_span=(t_start, t_end),
    y0=initial_state,
    t_eval=t_eval,
    args=(sigma, rho, beta),
    rtol=1e-9,
    atol=1e-12
)

# Extraction des composantes
t = solution.t
x = solution.y[0]
y = solution.y[1]
z = solution.y[2]

# Construction du DataFrame de trajectoire
lorenz_df = pd.DataFrame({
    "t": t,
    "x": x,
    "y": y,
    "z": z
})

lorenz_df.head()
```

4. Prétraitement et construction du vecteur d’état

Cette section prépare la trajectoire de Lorenz pour le calcul des observables GVH.

On construit le vecteur d’état :

[
\mathbf{X}(t) = (x(t),y(t),z(t)).
]

Chaque ligne du tableau représente donc l’état du système à un instant donné.

```python
# ============================================================
# Étape 4 — Prétraitement et construction du vecteur d’état
# ============================================================

# Extraction du vecteur d'état tridimensionnel
X = lorenz_df[["x", "y", "z"]].values

# Vérification de la forme du tableau
print("Forme de X :", X.shape)

# Aperçu des premiers états
X[:5]
```

5. Calcul de (D_T)

On définit (D_T) comme la distance locale entre deux états successifs du système :

[
D_T(t_i)=|\mathbf{X}(t_{i+1})-\mathbf{X}(t_i)|.
]

Pour le système de Lorenz, cette quantité mesure l’amplitude locale du déplacement dans l’espace des phases ((x,y,z)).

```python
# ============================================================
# Étape 5 — Calcul de D_T
# ============================================================

# Différences entre états successifs
dX = np.diff(X, axis=0)

# Distance locale entre états successifs
D_T = np.linalg.norm(dX, axis=1)

# Statistique principale
D_T_mean = np.mean(D_T)

print("D_T_mean =", D_T_mean)

# Aperçu
D_T[:10]
```

6. Calcul de (S_T) et (S_T^{norm})

On définit (S_T) comme la dispersion des distances locales (D_T) :

[
S_T = \sigma(D_T).
]

On définit aussi une version normalisée :

[
S_T^{norm} = \frac{S_T}{D_T^{mean}}.
]

Cette normalisation permet de comparer la variabilité relative entre plusieurs systèmes dynamiques.

```python
# ============================================================
# Étape 6 — Calcul de S_T et S_T_norm
# ============================================================

# Dispersion des distances locales
S_T = np.std(D_T)

# Dispersion normalisée
S_T_norm = S_T / D_T_mean

print("S_T =", S_T)
print("S_T_norm =", S_T_norm)
```

7. Calcul de (D_c) et (R_{\rm super})

On introduit un seuil critique (D_c) construit à partir de la distribution des distances locales (D_T).

Dans cette version 0.1, on définit :

[
D_c = \langle D_T \rangle + \sigma(D_T).
]

Le rapport supercritique est ensuite défini comme la fraction des valeurs vérifiant :

[
D_T > D_c.
]

Ainsi :

[
R_{\rm super}

\frac{N(D_T>D_c)}{N_{\rm total}}.
]

Cette quantité mesure la proportion des déplacements locaux situés au-dessus du seuil critique choisi.

```python
# ============================================================
# Étape 7 — Calcul de D_c et R_super
# ============================================================

# Seuil critique
D_c = D_T_mean + S_T

# Fraction des distances locales au-dessus du seuil
R_super = np.mean(D_T > D_c)

print("D_c =", D_c)
print("R_super =", R_super)
```

```python
8. Calcul de (A_H) et (B_H)

Pour caractériser la décroissance du rapport supercritique, on calcule (R_{\rm super}) pour une famille de seuils critiques (D_c).

On ajuste ensuite une loi exponentielle :

[
R_{\rm super}(D_c)=A_H e^{-B_H D_c}.
]

Les paramètres (A_H) et (B_H) résument la vitesse de décroissance de la fraction supercritique lorsque le seuil augmente.
```

```python
# ============================================================
# Étape 8 — Calcul de A_H et B_H
# ============================================================

# Famille de seuils critiques
D_c_values = np.linspace(np.min(D_T), np.max(D_T), 100)

# Calcul de R_super pour chaque seuil
R_super_values = np.array([
    np.mean(D_T > Dc) for Dc in D_c_values
])

# On retire les valeurs nulles pour permettre l'ajustement exponentiel
mask = R_super_values > 0

D_c_fit = D_c_values[mask]
R_super_fit = R_super_values[mask]

# Modèle exponentiel
def exponential_decay(Dc, A_H, B_H):
    return A_H * np.exp(-B_H * Dc)

# Ajustement
popt, pcov = curve_fit(
    exponential_decay,
    D_c_fit,
    R_super_fit,
    p0=(1.0, 1.0),
    maxfev=10000
)

A_H, B_H = popt

print("A_H =", A_H)
print("B_H =", B_H)
```

9. Calcul de (\theta_{\rm mean}), (\theta_{\max}) et (R_{180})

Cette section mesure la géométrie angulaire locale de la trajectoire.

On compare deux déplacements successifs :

[
\Delta \mathbf{X}i = \mathbf{X}{i+1}-\mathbf{X}_i
]

et

[
\Delta \mathbf{X}{i+1} = \mathbf{X}{i+2}-\mathbf{X}_{i+1}.
]

L’angle (\theta_i) entre ces deux déplacements est calculé par :

[
\cos(\theta_i)=
\frac{\Delta \mathbf{X}i\cdot \Delta \mathbf{X}{i+1}}
{|\Delta \mathbf{X}i||\Delta \mathbf{X}{i+1}|}.
]

On extrait ensuite :

* (\theta_{\rm mean}) : angle moyen ;
* (\theta_{\max}) : angle maximal ;
* (R_{180}) : fraction des angles proches de (180^\circ).

```python
# ============================================================
# Étape 9 — Calcul de theta_mean, theta_max et R_180
# ============================================================

# Vecteurs de déplacement successifs
V1 = dX[:-1]
V2 = dX[1:]

# Produits scalaires
dot_products = np.sum(V1 * V2, axis=1)

# Normes
norm_V1 = np.linalg.norm(V1, axis=1)
norm_V2 = np.linalg.norm(V2, axis=1)

# Cosinus des angles
cos_theta = dot_products / (norm_V1 * norm_V2)

# Sécurité numérique
cos_theta = np.clip(cos_theta, -1.0, 1.0)

# Angles en degrés
theta = np.degrees(np.arccos(cos_theta))

# Observables angulaires
theta_mean = np.mean(theta)
theta_max = np.max(theta)

# Seuil d'opposition angulaire
theta_threshold = 175.0
R_180 = np.mean(theta >= theta_threshold)

print("theta_mean =", theta_mean)
print("theta_max =", theta_max)
print("R_180 =", R_180)
```

```python
# ============================================================
# Étape 10 — Tableau récapitulatif des observables GVH
# ============================================================

lorenz_observables = pd.DataFrame({
    "Observable": [
        "D_T_mean",
        "S_T",
        "S_T_norm",
        "D_c",
        "R_super",
        "A_H",
        "B_H",
        "theta_mean",
        "theta_max",
        "R_180"
    ],
    "Valeur": [
        D_T_mean,
        S_T,
        S_T_norm,
        D_c,
        R_super,
        A_H,
        B_H,
        theta_mean,
        theta_max,
        R_180
    ]
})

lorenz_observables
```

```python
# ============================================================
# Étape 11 — Figures
# ============================================================

# Figure 1 — Trajectoire de Lorenz
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")
ax.plot(x, y, z, linewidth=0.5)
ax.set_title("Trajectoire de Lorenz")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.set_zlabel("z")
plt.show()

# Figure 2 — Distribution de D_T
plt.figure(figsize=(8, 5))
plt.hist(D_T, bins=50, alpha=0.8)
plt.axvline(D_c, linestyle="--", label="D_c")
plt.title("Distribution de D_T — Lorenz")
plt.xlabel("D_T")
plt.ylabel("Fréquence")
plt.legend()
plt.show()

# Figure 3 — Ajustement exponentiel de R_super(D_c)
plt.figure(figsize=(8, 5))
plt.scatter(D_c_fit, R_super_fit, s=15, label="Données")
plt.plot(
    D_c_fit,
    exponential_decay(D_c_fit, A_H, B_H),
    label="Fit exponentiel"
)
plt.title("Décroissance de R_super(D_c) — Lorenz")
plt.xlabel("D_c")
plt.ylabel("R_super")
plt.legend()
plt.show()

# Figure 4 — Distribution angulaire
plt.figure(figsize=(8, 5))
plt.hist(theta, bins=50, alpha=0.8)
plt.axvline(theta_threshold, linestyle="--", label="Seuil 175°")
plt.title("Distribution des angles locaux — Lorenz")
plt.xlabel("Angle θ (degrés)")
plt.ylabel("Fréquence")
plt.legend()
plt.show()
```

```python
# ============================================================
# Étape 12 — Export CSV
# ============================================================

# Export du tableau des observables GVH
lorenz_observables.to_csv(
    "GVH_Lorenz_Observables_0.1.csv",
    index=False
)

# Export de la trajectoire Lorenz
lorenz_df.to_csv(
    "GVH_Lorenz_Trajectory_0.1.csv",
    index=False
)

print("Exports terminés :")
print("- GVH_Lorenz_Observables_0.1.csv")
print("- GVH_Lorenz_Trajectory_0.1.csv")
```
