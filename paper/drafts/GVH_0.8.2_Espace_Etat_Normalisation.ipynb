## Théorème 2 — Séparation

La distance est nulle si et seulement si les deux états sont identiques.

$begin:math:display$
d\_\{\\mathcal X\}\(X\,Y\)\=0
\\Longleftrightarrow
X\=Y\.
$end:math:display$

### Démonstration

La somme de carrés ne peut être nulle que si

$begin:math:display$
\\Delta\\widetilde\{\\mathbf r\}\=0
$end:math:display$

et

$begin:math:display$
\\Delta\\widetilde\{\\mathbf v\}\=0\.
$end:math:display$

Les deux états sont donc identiques.

CQFD.

## Théorème 3 — Symétrie

Pour tous

$begin:math:display$
X\,Y\,
$end:math:display$

on obtient

$begin:math:display$
d\_\{\\mathcal X\}\(X\,Y\)
\=
d\_\{\\mathcal X\}\(Y\,X\)\.
$end:math:display$

### Démonstration

Changer l'ordre des états change simplement le signe des différences,

mais

$begin:math:display$
\(\-a\)\^2\=a\^2\.
$end:math:display$

La valeur de la métrique reste donc inchangée.

CQFD.

## Théorème 4 — Inégalité triangulaire

Pour tous

$begin:math:display$
X\,Y\,Z\,
$end:math:display$

on a

$begin:math:display$
d\(X\,Z\)
\\le
d\(X\,Y\)\+d\(Y\,Z\)\.
$end:math:display$

### Justification

La métrique GVH est construite à partir d'une norme euclidienne
dans l'espace d'état normalisé.

Toute norme euclidienne satisfait naturellement l'inégalité triangulaire.

Par conséquent,

$begin:math:display$
d\_\{\\mathcal X\}
$end:math:display$

est bien une métrique.

## Corollaire

La fonction

$begin:math:display$
d\_\{\\mathcal X\}
$end:math:display$

définit une structure métrique sur

$begin:math:display$
\\mathcal X\_N\.
$end:math:display$

Le couple

$begin:math:display$
\(\\mathcal X\_N\,d\_\{\\mathcal X\}\)
$end:math:display$

constitue un espace métrique.

Toutes les observables GVH pourront désormais être définies sur cet
espace métrique.

# 9. Opérateur de transition dynamique

## 9.1 Motivation

Jusqu’à présent, le framework GVH mesure la distance entre deux états
successifs :

\[
D_T(k)
=
d_{\mathcal X}(X_k,X_{k+1}).
\]

Cette définition mesure l’amplitude du changement observé entre les
instants \(t_k\) et \(t_{k+1}\).

Cependant, une dynamique ne se limite pas à l’existence de deux états
différents.

Elle correspond à une règle d’évolution qui transforme un état initial
en un état ultérieur.

Nous introduisons donc une application de transition :

\[
T_k :
\mathcal X_N
\longrightarrow
\mathcal X_N,
\]

telle que :

\[
X_{k+1}
=
T_k(X_k).
\]

L’opérateur \(T_k\) représente l’évolution du système pendant
l’intervalle temporel :

\[
[t_k,t_{k+1}].
\]

Il dépend des lois physiques, des paramètres du système et,
éventuellement, du temps.

## 9.2 Systèmes autonomes et systèmes dépendant du temps

L'opérateur de transition dépend des lois qui gouvernent l'évolution du
système.

Selon la nature de ces lois, deux situations principales peuvent être
distinguées.

### Système autonome

Un système est dit autonome lorsque son évolution dépend uniquement de
son état courant.

Mathématiquement,

$begin:math:display$
\\frac\{dX\}\{dt\}
\=
F\(X\)\.
$end:math:display$

Le temps n'apparaît pas explicitement dans la fonction dynamique.

L'opérateur de transition s'écrit alors

$begin:math:display$
T \:
\\mathcal X\_N
\\longrightarrow
\\mathcal X\_N\,
$end:math:display$

avec

$begin:math:display$
X\_\{k\+1\}
\=
T\(X\_k\)\.
$end:math:display$

La même règle d'évolution est appliquée à chaque instant.

---

### Système dépendant explicitement du temps

Un système est dit non autonome lorsque son évolution dépend à la fois
de son état et du temps.

Mathématiquement,

$begin:math:display$
\\frac\{dX\}\{dt\}
\=
F\(X\,t\)\.
$end:math:display$

Dans ce cas, la règle d'évolution peut varier au cours du temps.

L'opérateur de transition devient

$begin:math:display$
T\_k \:
\\mathcal X\_N
\\longrightarrow
\\mathcal X\_N\,
$end:math:display$

et

$begin:math:display$
X\_\{k\+1\}
\=
T\_k\(X\_k\)\.
$end:math:display$

L'indice $begin:math:text$k$end:math:text$ indique que la transition peut être différente d'un pas
de temps à l'autre.

### Interprétation dans le framework GVH

Le framework GVH ne suppose pas que le système soit autonome.

Il est construit pour décrire des transitions dynamiques dans les deux
situations.

Par exemple :

- une orbite képlérienne idéale peut être décrite comme un système
  autonome ;

- un satellite soumis à une manœuvre orbitale devient un système
  dépendant explicitement du temps ;

- un problème gravitationnel à N corps avec une poussée appliquée à un
  instant donné est également un système non autonome.

Le choix de l'opérateur de transition dépend donc des lois physiques
considérées, tandis que les observables GVH restent définies sur
l'évolution des états dans l'espace d'état.

## 9.3 Transition d'état

L'opérateur de transition

$begin:math:display$
T\_k \:
\\mathcal X\_N
\\longrightarrow
\\mathcal X\_N
$end:math:display$

fait évoluer l'état du système entre deux instants successifs.

Pour un état

$begin:math:display$
X\_k\=X\(t\_k\)\,
$end:math:display$

on obtient

$begin:math:display$
X\_\{k\+1\}
\=
T\_k\(X\_k\)\.
$end:math:display$

La transition d'état est définie comme la différence entre
l'état final et l'état initial :

$begin:math:display$
\\boxed\{
\\Delta X\_k
\=
X\_\{k\+1\}
\-
X\_k
\}
$end:math:display$

En utilisant l'opérateur de transition,

$begin:math:display$
\\boxed\{
\\Delta X\_k
\=
T\_k\(X\_k\)\-X\_k\.
\}
$end:math:display$

La quantité

$begin:math:display$
\\Delta X\_k
$end:math:display$

représente l'évolution complète du système pendant
l'intervalle

$begin:math:display$
\[t\_k\,t\_\{k\+1\}\]\.
$end:math:display$

## 9.4 Nécessité d'une normalisation

La transition d'état

$begin:math:display$
\\Delta X\_k
\=
T\_k\(X\_k\)\-X\_k
$end:math:display$

contient simultanément :

- des positions, exprimées en mètres ;

- des vitesses, exprimées en mètres par seconde.

Ces grandeurs possèdent des dimensions physiques différentes.

Il n'est donc pas possible de définir directement une norme physique sur
$begin:math:text$\\Delta X\_k$end:math:text$.

Afin de construire une métrique cohérente, le framework GVH introduit un
opérateur de normalisation qui transforme toutes les composantes de
l'espace d'état en variables sans dimension.

## 9.5 Opérateur de normalisation

Soient

$begin:math:display$
L\_\\star\>0
$end:math:display$

une longueur caractéristique,

et

$begin:math:display$
V\_\\star\>0
$end:math:display$

une vitesse caractéristique.

Nous définissons l'opérateur

$begin:math:display$
\\mathcal N \:
\\mathcal X\_N
\\longrightarrow
\\widetilde\{\\mathcal X\}\_N\,
$end:math:display$

par

$begin:math:display$
\\widetilde X
\=
\\mathcal N\(X\)\.
$end:math:display$

Pour chaque corps,

$begin:math:display$
\\widetilde\{\\mathbf r\}
\=
\\frac\{\\mathbf r\}\{L\_\\star\}\,
$end:math:display$

et

$begin:math:display$
\\widetilde\{\\mathbf v\}
\=
\\frac\{\\mathbf v\}\{V\_\\star\}\.
$end:math:display$

Ainsi,

$begin:math:display$
\\widetilde X
\=
\(
\\widetilde\{\\mathbf r\}\_1\,
\\widetilde\{\\mathbf v\}\_1\,
\\ldots\,
\\widetilde\{\\mathbf r\}\_N\,
\\widetilde\{\\mathbf v\}\_N
\)\.
$end:math:display$

Toutes les composantes deviennent alors adimensionnées.

## 9.6 Transition normalisée

La transition d'état normalisée est définie par

$begin:math:display$
\\Delta\\widetilde X\_k
\=
\\widetilde X\_\{k\+1\}
\-
\\widetilde X\_k\.
$end:math:display$

En utilisant les opérateurs précédents,

$begin:math:display$
\\widetilde X\_k
\=
\\mathcal N\(X\_k\)\,
$end:math:display$

et

$begin:math:display$
X\_\{k\+1\}
\=
T\_k\(X\_k\)\,
$end:math:display$

on obtient

$begin:math:display$
\\boxed\{
\\Delta\\widetilde X\_k
\=
\\mathcal N\(T\_k\(X\_k\)\)
\-
\\mathcal N\(X\_k\)\.
\}
$end:math:display$

Cette quantité représente la transition dynamique exprimée dans un
espace d'état normalisé.

Elle constitue l'objet fondamental sur lequel seront construites les
observables GVH.

## 9.7 Norme de la transition dynamique

L’espace d’état normalisé

$begin:math:display$
\\widetilde\{\\mathcal X\}\_N
\\simeq
\\mathbb R\^\{6N\}
$end:math:display$

est muni de la norme euclidienne.

Pour tout vecteur normalisé

$begin:math:display$
Y
\=
\(
\\mathbf y\_1\,\\ldots\,\\mathbf y\_\{6N\}
\)
\\in
\\widetilde\{\\mathcal X\}\_N\,
$end:math:display$

sa norme est définie par

$begin:math:display$
\\\|Y\\\|\_\{\\widetilde\{\\mathcal X\}\}
\=
\\left\(
\\sum\_\{j\=1\}\^\{6N\} y\_j\^2
\\right\)\^\{1\/2\}\.
$end:math:display$

La transition normalisée entre les instants $begin:math:text$t\_k$end:math:text$ et $begin:math:text$t\_\{k\+1\}$end:math:text$ est

$begin:math:display$
\\Delta\\widetilde X\_k
\=
\\widetilde X\_\{k\+1\}
\-
\\widetilde X\_k\.
$end:math:display$

La première observable dynamique du framework GVH est alors définie par

$begin:math:display$
\\boxed\{
D\_T\(k\)
\=
\\left\\\|
\\Delta\\widetilde X\_k
\\right\\\|\_\{\\widetilde\{\\mathcal X\}\}
\}
$end:math:display$

ou, de manière équivalente,

$begin:math:display$
\\boxed\{
D\_T\(k\)
\=
\\left\\\|
\\mathcal N\(T\_k\(X\_k\)\)
\-
\\mathcal N\(X\_k\)
\\right\\\|\_\{\\widetilde\{\\mathcal X\}\}\.
\}
$end:math:display$

La quantité $begin:math:text$D\_T\(k\)$end:math:text$ mesure l’amplitude géométrique sans dimension de
la transition produite pendant l’intervalle

$begin:math:display$
\[t\_k\,t\_\{k\+1\}\]\.
$end:math:display$

### Expression explicite pour un système à $begin:math:text$N$end:math:text$ corps

Pour chaque corps $begin:math:text$i$end:math:text$, la transition normalisée contient une variation
de position et une variation de vitesse :

$begin:math:display$
\\Delta\\widetilde\{\\mathbf r\}\_\{i\,k\}
\=
\\frac\{
\\mathbf r\_\{i\,k\+1\}\-\\mathbf r\_\{i\,k\}
\}\{
L\_\\star
\}\,
$end:math:display$

$begin:math:display$
\\Delta\\widetilde\{\\mathbf v\}\_\{i\,k\}
\=
\\frac\{
\\mathbf v\_\{i\,k\+1\}\-\\mathbf v\_\{i\,k\}
\}\{
V\_\\star
\}\.
$end:math:display$

La norme globale de la transition est donc

$begin:math:display$
\\boxed\{
D\_T\(k\)
\=
\\left\[
\\sum\_\{i\=1\}\^\{N\}
\\left\(
\\left\\\|
\\Delta\\widetilde\{\\mathbf r\}\_\{i\,k\}
\\right\\\|\_2\^2
\+
\\left\\\|
\\Delta\\widetilde\{\\mathbf v\}\_\{i\,k\}
\\right\\\|\_2\^2
\\right\)
\\right\]\^\{1\/2\}\.
\}
$end:math:display$

De manière équivalente,

$begin:math:display$
D\_T\(k\)
\=
\\left\[
\\sum\_\{i\=1\}\^\{N\}
\\left\(
\\frac\{
\\\|\\mathbf r\_\{i\,k\+1\}\-\\mathbf r\_\{i\,k\}\\\|\_2\^2
\}\{
L\_\\star\^2
\}
\+
\\frac\{
\\\|\\mathbf v\_\{i\,k\+1\}\-\\mathbf v\_\{i\,k\}\\\|\_2\^2
\}\{
V\_\\star\^2
\}
\\right\)
\\right\]\^\{1\/2\}\.
$end:math:display$

Ainsi, $begin:math:text$D\_T\(k\)$end:math:text$ combine les variations de position et de vitesse dans
une même quantité adimensionnée.

## 9.8 Compatibilité avec la métrique de l’espace d’état

La métrique de l’espace d’état normalisé est définie par

$begin:math:display$
d\_\{\\widetilde\{\\mathcal X\}\}
\(
\\widetilde X\,
\\widetilde Y
\)
\=
\\\|
\\widetilde Y\-\\widetilde X
\\\|\_\{\\widetilde\{\\mathcal X\}\}\.
$end:math:display$

En choisissant

$begin:math:display$
\\widetilde X\=\\widetilde X\_k
$end:math:display$

et

$begin:math:display$
\\widetilde Y\=\\widetilde X\_\{k\+1\}\,
$end:math:display$

on obtient

$begin:math:display$
d\_\{\\widetilde\{\\mathcal X\}\}
\(
\\widetilde X\_k\,
\\widetilde X\_\{k\+1\}
\)
\=
\\\|
\\widetilde X\_\{k\+1\}
\-
\\widetilde X\_k
\\\|\_\{\\widetilde\{\\mathcal X\}\}\.
$end:math:display$

Or,

$begin:math:display$
\\Delta\\widetilde X\_k
\=
\\widetilde X\_\{k\+1\}
\-
\\widetilde X\_k\.
$end:math:display$

Par conséquent,

$begin:math:display$
\\boxed\{
D\_T\(k\)
\=
\\\|
\\Delta\\widetilde X\_k
\\\|\_\{\\widetilde\{\\mathcal X\}\}
\=
d\_\{\\widetilde\{\\mathcal X\}\}
\(
\\widetilde X\_k\,
\\widetilde X\_\{k\+1\}
\)\.
\}
$end:math:display$

La définition dynamique de $begin:math:text$D\_T$end:math:text$ et sa définition métrique sont donc
équivalentes.

### Proposition 1 — Équivalence métrique et dynamique de $begin:math:text$D\_T$end:math:text$

Soit

$begin:math:display$
T\_k\:
\\mathcal X\_N
\\longrightarrow
\\mathcal X\_N
$end:math:display$

un opérateur de transition, et soit

$begin:math:display$
\\mathcal N\:
\\mathcal X\_N
\\longrightarrow
\\widetilde\{\\mathcal X\}\_N
$end:math:display$

un opérateur de normalisation.

Pour tout état $begin:math:text$X\_k\\in\\mathcal X\_N$end:math:text$, on définit

$begin:math:display$
\\Delta\\widetilde X\_k
\=
\\mathcal N\(T\_k\(X\_k\)\)
\-
\\mathcal N\(X\_k\)\.
$end:math:display$

Alors,

$begin:math:display$
D\_T\(k\)
\=
\\\|
\\Delta\\widetilde X\_k
\\\|\_\{\\widetilde\{\\mathcal X\}\}
$end:math:display$

est égal à la distance entre l’état normalisé initial et l’état
normalisé final :

$begin:math:display$
\\boxed\{
D\_T\(k\)
\=
d\_\{\\widetilde\{\\mathcal X\}\}
\\left\(
\\mathcal N\(X\_k\)\,
\\mathcal N\(T\_k\(X\_k\)\)
\\right\)\.
\}
$end:math:display$

#### Démonstration

Par définition de la métrique induite par la norme,

$begin:math:display$
d\_\{\\widetilde\{\\mathcal X\}\}
\(
\\widetilde X\,
\\widetilde Y
\)
\=
\\\|
\\widetilde Y\-\\widetilde X
\\\|\_\{\\widetilde\{\\mathcal X\}\}\.
$end:math:display$

En posant

$begin:math:display$
\\widetilde X
\=
\\mathcal N\(X\_k\)
$end:math:display$

et

$begin:math:display$
\\widetilde Y
\=
\\mathcal N\(T\_k\(X\_k\)\)\,
$end:math:display$

il vient

$begin:math:display$
d\_\{\\widetilde\{\\mathcal X\}\}
\\left\(
\\mathcal N\(X\_k\)\,
\\mathcal N\(T\_k\(X\_k\)\)
\\right\)
\=
\\left\\\|
\\mathcal N\(T\_k\(X\_k\)\)
\-
\\mathcal N\(X\_k\)
\\right\\\|\_\{\\widetilde\{\\mathcal X\}\}\.
$end:math:display$

Le membre de droite est précisément

$begin:math:display$
\\\|
\\Delta\\widetilde X\_k
\\\|\_\{\\widetilde\{\\mathcal X\}\}\.
$end:math:display$

Par conséquent,

$begin:math:display$
D\_T\(k\)
\=
d\_\{\\widetilde\{\\mathcal X\}\}
\\left\(
\\mathcal N\(X\_k\)\,
\\mathcal N\(T\_k\(X\_k\)\)
\\right\)\.
$end:math:display$

$begin:math:display$
\\blacksquare
$end:math:display$

```python
# ============================================================
# Chargement robuste des fichiers JPL Horizons RAW
# Jupiter — Io — Europa — Ganymede
# ============================================================

from pathlib import Path
from io import StringIO
import pandas as pd
import csv
import re


# ------------------------------------------------------------
# 1. Fichiers attendus
# ------------------------------------------------------------

fichiers_horizons = {
    "Io": Path("Io_Horizons_Jovicentric_Vector_2026_1h_RAW.csv"),
    "Europa": Path("Europa_Horizons_Jovicentric_Vector_2026_1h_RAW.csv"),
    "Ganymede": Path("Ganymede_Horizons_Jovicentric_Vector_2026_1h_RAW.csv")
}


# ------------------------------------------------------------
# 2. Fonction de nettoyage des noms de colonnes
# ------------------------------------------------------------

def nettoyer_nom_colonne(nom):
    """
    Nettoie un nom de colonne provenant de JPL Horizons.
    """
    nom = str(nom).strip()
    nom = nom.replace('"', "")
    nom = nom.replace(" ", "_")
    nom = nom.replace("(", "")
    nom = nom.replace(")", "")
    nom = nom.replace("/", "_")
    nom = re.sub(r"_+", "_", nom)

    return nom.strip("_")


# ------------------------------------------------------------
# 3. Fonction de lecture d'un fichier Horizons RAW
# ------------------------------------------------------------

def lire_horizons_raw(chemin):
    """
    Lit un fichier brut JPL Horizons.

    La fonction :
    - localise $$SOE et $$EOE ;
    - extrait uniquement les données ;
    - récupère l'en-tête situé avant $$SOE ;
    - supprime les colonnes entièrement vides ;
    - convertit les colonnes numériques.
    """

    chemin = Path(chemin)

    if not chemin.exists():
        raise FileNotFoundError(
            f"Fichier introuvable : {chemin}"
        )

    # Lecture texte robuste
    texte = chemin.read_text(
        encoding="utf-8",
        errors="replace"
    )

    lignes = texte.splitlines()

    # Localisation des marqueurs Horizons
    indices_soe = [
        i for i, ligne in enumerate(lignes)
        if "$$SOE" in ligne
    ]

    indices_eoe = [
        i for i, ligne in enumerate(lignes)
        if "$$EOE" in ligne
    ]

    if not indices_soe:
        raise ValueError(
            f"Marqueur $$SOE absent dans {chemin.name}"
        )

    if not indices_eoe:
        raise ValueError(
            f"Marqueur $$EOE absent dans {chemin.name}"
        )

    debut = indices_soe[0]
    fin = next(
        i for i in indices_eoe
        if i > debut
    )

    # --------------------------------------------------------
    # Recherche de l'en-tête CSV avant $$SOE
    # --------------------------------------------------------

    ligne_entete = None

    for i in range(debut - 1, max(-1, debut - 30), -1):
        candidate = lignes[i].strip()

        if (
            "," in candidate
            and not candidate.startswith("*")
            and not candidate.startswith("!")
            and not candidate.startswith("$$")
        ):
            ligne_entete = candidate
            break

    # --------------------------------------------------------
    # Extraction des lignes de données
    # --------------------------------------------------------

    lignes_donnees = [
        ligne.strip()
        for ligne in lignes[debut + 1:fin]
        if ligne.strip()
    ]

    if not lignes_donnees:
        raise ValueError(
            f"Aucune donnée entre $$SOE et $$EOE dans {chemin.name}"
        )

    # Suppression des virgules terminales éventuelles
    lignes_donnees = [
        ligne.rstrip(",")
        for ligne in lignes_donnees
    ]

    # --------------------------------------------------------
    # Lecture avec le module csv
    # --------------------------------------------------------

    lignes_parsees = list(
        csv.reader(
            lignes_donnees,
            skipinitialspace=True
        )
    )

    nombre_colonnes = max(
        len(ligne) for ligne in lignes_parsees
    )

    # Uniformisation du nombre de colonnes
    lignes_parsees = [
        ligne + [""] * (nombre_colonnes - len(ligne))
        for ligne in lignes_parsees
    ]

    # --------------------------------------------------------
    # Détermination des noms de colonnes
    # --------------------------------------------------------

    if ligne_entete is not None:
        colonnes = next(
            csv.reader(
                [ligne_entete],
                skipinitialspace=True
            )
        )

        colonnes = [
            nettoyer_nom_colonne(c)
            for c in colonnes
        ]

    else:
        colonnes = []

    # Ajustement si l'en-tête ne correspond pas exactement
    if len(colonnes) < nombre_colonnes:
        colonnes += [
            f"colonne_{i + 1}"
            for i in range(
                len(colonnes),
                nombre_colonnes
            )
        ]

    elif len(colonnes) > nombre_colonnes:
        colonnes = colonnes[:nombre_colonnes]

    # Noms vides ou dupliqués
    colonnes_finales = []
    compteurs = {}

    for i, colonne in enumerate(colonnes):
        if not colonne:
            colonne = f"colonne_{i + 1}"

        if colonne in compteurs:
            compteurs[colonne] += 1
            colonne = f"{colonne}_{compteurs[colonne]}"
        else:
            compteurs[colonne] = 1

        colonnes_finales.append(colonne)

    # --------------------------------------------------------
    # Construction du DataFrame
    # --------------------------------------------------------

    df = pd.DataFrame(
        lignes_parsees,
        columns=colonnes_finales
    )

    # Nettoyage des espaces
    for colonne in df.columns:
        if df[colonne].dtype == "object":
            df[colonne] = df[colonne].astype(str).str.strip()

    # Remplacement des chaînes vides
    df = df.replace(
        {
            "": pd.NA,
            "n.a.": pd.NA,
            "N.A.": pd.NA,
            "nan": pd.NA
        }
    )

    # Suppression des colonnes totalement vides
    df = df.dropna(
        axis=1,
        how="all"
    )

    # --------------------------------------------------------
    # Conversion automatique des colonnes numériques
    # --------------------------------------------------------

    for colonne in df.columns:
        serie_numerique = pd.to_numeric(
            df[colonne],
            errors="coerce"
        )

        proportion_numerique = serie_numerique.notna().mean()

        # Conversion si au moins 90 % des valeurs sont numériques
        if proportion_numerique >= 0.90:
            df[colonne] = serie_numerique

    return df


# ------------------------------------------------------------
# 4. Chargement des trois satellites
# ------------------------------------------------------------

donnees = {}

print("Chargement des fichiers Horizons RAW...\n")

for satellite, chemin in fichiers_horizons.items():

    try:
        df = lire_horizons_raw(chemin)

        donnees[satellite] = df

        print(
            f"{satellite:10s} : "
            f"{df.shape[0]} lignes × "
            f"{df.shape[1]} colonnes"
        )

    except Exception as erreur:
        print(f"{satellite:10s} : ERREUR")
        print(f"  {erreur}")


# ------------------------------------------------------------
# 5. Inspection des colonnes et premières lignes
# ------------------------------------------------------------

for satellite, df in donnees.items():

    print("\n" + "=" * 80)
    print(satellite)
    print("=" * 80)

    print("\nColonnes détectées :")

    for i, colonne in enumerate(df.columns, start=1):
        print(
            f"{i:2d}. {colonne} "
            f"— type : {df[colonne].dtype}"
        )

    print("\nPremières lignes :")
    display(df.head(3))


# ------------------------------------------------------------
# 6. Vérification des dimensions
# ------------------------------------------------------------

print("\nRésumé des dimensions :\n")

for satellite in ["Io", "Europa", "Ganymede"]:

    if satellite in donnees:
        df = donnees[satellite]

        print(
            f"{satellite:10s} : "
            f"{len(df)} états temporels"
        )

    else:
        print(
            f"{satellite:10s} : non chargé"
        )


# ------------------------------------------------------------
# 7. Recherche des colonnes communes exactes
# ------------------------------------------------------------

if len(donnees) == 3:

    colonnes_communes = sorted(
        set(donnees["Io"].columns)
        & set(donnees["Europa"].columns)
        & set(donnees["Ganymede"].columns)
    )

    print("\nColonnes communes aux trois satellites :\n")

    for colonne in colonnes_communes:
        print("-", colonne)

else:
    colonnes_communes = []

    print(
        "\nLes trois fichiers doivent être chargés "
        "avant de rechercher les colonnes communes."
    )


# ------------------------------------------------------------
# 8. Colonnes numériques communes
# ------------------------------------------------------------

colonnes_numeriques_communes = []

if len(donnees) == 3:

    for colonne in colonnes_communes:

        est_numerique_partout = all(
            pd.api.types.is_numeric_dtype(
                donnees[satellite][colonne]
            )
            for satellite in [
                "Io",
                "Europa",
                "Ganymede"
            ]
        )

        if est_numerique_partout:
            colonnes_numeriques_communes.append(colonne)

    print("\nColonnes numériques communes :\n")

    for colonne in colonnes_numeriques_communes:
        print("-", colonne)
```

```python
# ============================================================
# Lecture harmonisée JPL Horizons
# Io : format texte multiligne
# Europa / Ganymede : format CSV tabulaire
# ============================================================

from pathlib import Path
import pandas as pd
import numpy as np
import re


# ------------------------------------------------------------
# 1. Fichiers
# ------------------------------------------------------------

fichiers = {
    "Io": Path("Io_Horizons_Jovicentric_Vector_2026_1h_RAW.csv"),
    "Europa": Path("Europa_Horizons_Jovicentric_Vector_2026_1h_RAW.csv"),
    "Ganymede": Path("Ganymede_Horizons_Jovicentric_Vector_2026_1h_RAW.csv")
}


# ------------------------------------------------------------
# 2. Fonction pour Europa et Ganymede
# ------------------------------------------------------------

def lire_horizons_csv_tabulaire(chemin):
    """
    Lit un export Horizons déjà organisé en colonnes CSV.
    """

    texte = Path(chemin).read_text(
        encoding="utf-8",
        errors="replace"
    )

    lignes = texte.splitlines()

    debut = next(
        i for i, ligne in enumerate(lignes)
        if "$$SOE" in ligne
    )

    fin = next(
        i for i, ligne in enumerate(lignes)
        if "$$EOE" in ligne and i > debut
    )

    # Recherche de l'en-tête avant $$SOE
    entete = None

    for i in range(debut - 1, max(-1, debut - 30), -1):
        ligne = lignes[i].strip()

        if (
            "JDTDB" in ligne
            and "Calendar Date" in ligne
            and "," in ligne
        ):
            entete = ligne
            break

    if entete is None:
        raise ValueError(
            f"En-tête tabulaire introuvable dans {chemin}"
        )

    donnees_brutes = [
        ligne.rstrip(",")
        for ligne in lignes[debut + 1:fin]
        if ligne.strip()
    ]

    bloc = "\n".join([entete] + donnees_brutes)

    from io import StringIO

    df = pd.read_csv(
        StringIO(bloc),
        skipinitialspace=True
    )

    # Nettoyage des noms
    df.columns = [
        str(c)
        .strip()
        .replace(" ", "_")
        .replace("(", "")
        .replace(")", "")
        for c in df.columns
    ]

    # Harmonisation attendue
    renommage = {
        "Calendar_Date_TDB": "Calendar_Date_TDB",
        "Calendar_Date_TDB_": "Calendar_Date_TDB"
    }

    df = df.rename(columns=renommage)

    colonnes = [
        "JDTDB",
        "Calendar_Date_TDB",
        "X", "Y", "Z",
        "VX", "VY", "VZ",
        "LT", "RG", "RR"
    ]

    manquantes = [
        c for c in colonnes
        if c not in df.columns
    ]

    if manquantes:
        raise ValueError(
            f"Colonnes absentes dans {chemin.name} : {manquantes}"
        )

    df = df[colonnes].copy()

    for colonne in [
        "JDTDB",
        "X", "Y", "Z",
        "VX", "VY", "VZ",
        "LT", "RG", "RR"
    ]:
        df[colonne] = pd.to_numeric(
            df[colonne],
            errors="coerce"
        )

    return df


# ------------------------------------------------------------
# 3. Fonction spéciale pour Io
# ------------------------------------------------------------

def extraire_valeur(ligne, variable):
    """
    Extrait une valeur scientifique telle que :
    X = 3.7118E+05
    VX= 8.3120E+00
    """

    motif = rf"\b{variable}\s*=\s*([+-]?\d+(?:\.\d*)?(?:[Ee][+-]?\d+)?)"

    resultat = re.search(
        motif,
        ligne
    )

    if resultat is None:
        return np.nan

    return float(resultat.group(1))


def lire_io_multiligne(chemin):
    """
    Lit le fichier Io Horizons où chaque époque est répartie
    sur quatre lignes :
    date, position, vitesse, grandeurs dérivées.
    """

    texte = Path(chemin).read_text(
        encoding="utf-8",
        errors="replace"
    )

    lignes = texte.splitlines()

    debut = next(
        i for i, ligne in enumerate(lignes)
        if "$$SOE" in ligne
    )

    fin = next(
        i for i, ligne in enumerate(lignes)
        if "$$EOE" in ligne and i > debut
    )

    lignes_donnees = [
        ligne.strip()
        for ligne in lignes[debut + 1:fin]
        if ligne.strip()
    ]

    observations = []
    i = 0

    motif_date = re.compile(
        r"^\s*"
        r"([0-9]+\.[0-9]+)"
        r"\s*=\s*"
        r"(.+?)"
        r"\s*$"
    )

    while i < len(lignes_donnees):

        ligne_date = lignes_donnees[i]
        resultat_date = motif_date.match(ligne_date)

        if resultat_date is None:
            i += 1
            continue

        if i + 3 >= len(lignes_donnees):
            break

        ligne_position = lignes_donnees[i + 1]
        ligne_vitesse = lignes_donnees[i + 2]
        ligne_derivees = lignes_donnees[i + 3]

        observation = {
            "JDTDB": float(resultat_date.group(1)),
            "Calendar_Date_TDB": resultat_date.group(2).strip(),

            "X": extraire_valeur(
                ligne_position, "X"
            ),
            "Y": extraire_valeur(
                ligne_position, "Y"
            ),
            "Z": extraire_valeur(
                ligne_position, "Z"
            ),

            "VX": extraire_valeur(
                ligne_vitesse, "VX"
            ),
            "VY": extraire_valeur(
                ligne_vitesse, "VY"
            ),
            "VZ": extraire_valeur(
                ligne_vitesse, "VZ"
            ),

            "LT": extraire_valeur(
                ligne_derivees, "LT"
            ),
            "RG": extraire_valeur(
                ligne_derivees, "RG"
            ),
            "RR": extraire_valeur(
                ligne_derivees, "RR"
            )
        }

        observations.append(observation)

        i += 4

    df = pd.DataFrame(observations)

    colonnes = [
        "JDTDB",
        "Calendar_Date_TDB",
        "X", "Y", "Z",
        "VX", "VY", "VZ",
        "LT", "RG", "RR"
    ]

    if df.empty:
        raise ValueError(
            "Aucune observation Io n'a été reconstruite."
        )

    return df[colonnes]


# ------------------------------------------------------------
# 4. Chargement harmonisé
# ------------------------------------------------------------

donnees = {}

donnees["Io"] = lire_io_multiligne(
    fichiers["Io"]
)

donnees["Europa"] = lire_horizons_csv_tabulaire(
    fichiers["Europa"]
)

donnees["Ganymede"] = lire_horizons_csv_tabulaire(
    fichiers["Ganymede"]
)


# ------------------------------------------------------------
# 5. Résumé
# ------------------------------------------------------------

print("Chargement harmonisé terminé :\n")

for satellite, df in donnees.items():
    print(
        f"{satellite:10s} : "
        f"{df.shape[0]} lignes × "
        f"{df.shape[1]} colonnes"
    )


# ------------------------------------------------------------
# 6. Vérification des colonnes
# ------------------------------------------------------------

colonnes_attendues = [
    "JDTDB",
    "Calendar_Date_TDB",
    "X", "Y", "Z",
    "VX", "VY", "VZ",
    "LT", "RG", "RR"
]

for satellite, df in donnees.items():

    print("\n" + "=" * 75)
    print(satellite)
    print("=" * 75)

    print("Colonnes :", list(df.columns))

    print("\nValeurs manquantes :")
    print(df.isna().sum())

    print("\nPremières lignes :")
    display(df.head(3))


# ------------------------------------------------------------
# 7. Vérification des colonnes communes
# ------------------------------------------------------------

colonnes_communes = sorted(
    set(donnees["Io"].columns)
    & set(donnees["Europa"].columns)
    & set(donnees["Ganymede"].columns)
)

print("\nColonnes communes :\n")

for colonne in colonnes_communes:
    print("-", colonne)


# ------------------------------------------------------------
# 8. Vérification des dimensions et dates
# ------------------------------------------------------------

resume = []

for satellite, df in donnees.items():

    resume.append({
        "Satellite": satellite,
        "N_lignes": len(df),
        "JDTDB_min": df["JDTDB"].min(),
        "JDTDB_max": df["JDTDB"].max(),
        "Doublons_JDTDB": int(
            df["JDTDB"].duplicated().sum()
        ),
        "NaN_etat": int(
            df[
                ["X", "Y", "Z", "VX", "VY", "VZ"]
            ].isna().sum().sum()
        )
    })

resume_df = pd.DataFrame(resume)

print("\nAudit temporel :")
display(resume_df)
```

```python
# ============================================================
# Construction de l'état collectif X_k dans R^18
# Io — Europa — Ganymede
# ============================================================

import numpy as np
import pandas as pd


# ------------------------------------------------------------
# 1. Variables primitives de l'espace d'état
# ------------------------------------------------------------

variables_etat = [
    "X", "Y", "Z",
    "VX", "VY", "VZ"
]

satellites = [
    "Io",
    "Europa",
    "Ganymede"
]


# ------------------------------------------------------------
# 2. Vérification stricte de l'alignement temporel
# ------------------------------------------------------------

temps_reference = donnees["Io"]["JDTDB"].to_numpy(
    dtype=float
)

for satellite in ["Europa", "Ganymede"]:

    temps_satellite = donnees[satellite]["JDTDB"].to_numpy(
        dtype=float
    )

    if not np.array_equal(
        temps_reference,
        temps_satellite
    ):
        raise ValueError(
            f"Les temps de {satellite} ne sont pas "
            f"strictement alignés avec ceux de Io."
        )

print("Alignement temporel strict : VALIDÉ")
print(f"Nombre d'états communs : {len(temps_reference)}")


# ------------------------------------------------------------
# 3. Extraction des états individuels
# ------------------------------------------------------------

etats_individuels = {}

for satellite in satellites:

    matrice = donnees[satellite][
        variables_etat
    ].to_numpy(dtype=float)

    etats_individuels[satellite] = matrice

    print(
        f"{satellite:10s} : "
        f"{matrice.shape}"
    )


# ------------------------------------------------------------
# 4. Construction de l'état collectif
# ------------------------------------------------------------

X_collectif = np.hstack([
    etats_individuels["Io"],
    etats_individuels["Europa"],
    etats_individuels["Ganymede"]
])

print("\nÉtat collectif construit :")
print("Forme de X_collectif :", X_collectif.shape)


# ------------------------------------------------------------
# 5. Noms explicites des 18 composantes
# ------------------------------------------------------------

colonnes_etat = []

for satellite in satellites:
    for variable in variables_etat:
        colonnes_etat.append(
            f"{satellite}_{variable}"
        )

X_collectif_df = pd.DataFrame(
    X_collectif,
    columns=colonnes_etat
)

X_collectif_df.insert(
    0,
    "Calendar_Date_TDB",
    donnees["Io"]["Calendar_Date_TDB"].to_numpy()
)

X_collectif_df.insert(
    0,
    "JDTDB",
    temps_reference
)


# ------------------------------------------------------------
# 6. Contrôles structurels
# ------------------------------------------------------------

assert X_collectif.shape == (
    len(temps_reference),
    18
)

assert not np.isnan(X_collectif).any()

assert np.isfinite(X_collectif).all()

print("\nContrôles de l'état collectif :")
print("- Dimension attendue R^18 : VALIDÉ")
print("- Absence de NaN : VALIDÉ")
print("- Valeurs finies : VALIDÉ")


# ------------------------------------------------------------
# 7. Aperçu
# ------------------------------------------------------------

print("\nPremiers états collectifs :")
display(X_collectif_df.head(3))
```

```python
# ============================================================
# Transition d'état collective
# Delta X_k = X_{k+1} - X_k
# ============================================================

import numpy as np
import pandas as pd


# ------------------------------------------------------------
# 1. Calcul de la transition brute
# ------------------------------------------------------------

Delta_X_collectif = np.diff(
    X_collectif,
    axis=0
)

print("Transition collective calculée.")
print("Forme de Delta_X_collectif :", Delta_X_collectif.shape)


# ------------------------------------------------------------
# 2. Temps associés aux transitions
# ------------------------------------------------------------

temps_transition = temps_reference[:-1]

dates_transition = donnees["Io"][
    "Calendar_Date_TDB"
].to_numpy()[:-1]


# ------------------------------------------------------------
# 3. Noms des composantes de transition
# ------------------------------------------------------------

colonnes_transition = [
    f"Delta_{colonne}"
    for colonne in colonnes_etat
]


# ------------------------------------------------------------
# 4. DataFrame des transitions
# ------------------------------------------------------------

Delta_X_collectif_df = pd.DataFrame(
    Delta_X_collectif,
    columns=colonnes_transition
)

Delta_X_collectif_df.insert(
    0,
    "Calendar_Date_TDB",
    dates_transition
)

Delta_X_collectif_df.insert(
    0,
    "JDTDB",
    temps_transition
)


# ------------------------------------------------------------
# 5. Contrôles structurels
# ------------------------------------------------------------

assert Delta_X_collectif.shape == (
    len(X_collectif) - 1,
    18
)

assert not np.isnan(
    Delta_X_collectif
).any()

assert np.isfinite(
    Delta_X_collectif
).all()

print("\nContrôles de la transition :")
print("- Nombre de transitions attendu : 8760")
print(
    "- Nombre obtenu :",
    Delta_X_collectif.shape[0]
)
print("- Dimension de chaque transition : 18")
print("- Absence de NaN : VALIDÉ")
print("- Valeurs finies : VALIDÉ")


# ------------------------------------------------------------
# 6. Vérification directe de la définition
# ------------------------------------------------------------

verification = np.allclose(
    Delta_X_collectif[0],
    X_collectif[1] - X_collectif[0]
)

print(
    "\nVérification "
    "Delta X_0 = X_1 - X_0 :",
    verification
)


# ------------------------------------------------------------
# 7. Aperçu
# ------------------------------------------------------------

print("\nPremières transitions collectives :")
display(
    Delta_X_collectif_df.head(3)
)
```

```python
# ============================================================
# Normalisation de l'état collectif dans R^18
# puis calcul de Delta X_tilde
# ============================================================

import numpy as np
import pandas as pd


# ------------------------------------------------------------
# 1. Paramètres de normalisation
# ------------------------------------------------------------

moyenne_etat = np.mean(
    X_collectif,
    axis=0
)

ecart_type_etat = np.std(
    X_collectif,
    axis=0,
    ddof=0
)


# ------------------------------------------------------------
# 2. Vérification des échelles non nulles
# ------------------------------------------------------------

indices_ecart_nul = np.where(
    ecart_type_etat == 0
)[0]

if len(indices_ecart_nul) > 0:

    composantes_problematiques = [
        colonnes_etat[i]
        for i in indices_ecart_nul
    ]

    raise ValueError(
        "Écart-type nul pour les composantes suivantes : "
        f"{composantes_problematiques}"
    )

print("Tous les écarts-types sont strictement positifs : VALIDÉ")


# ------------------------------------------------------------
# 3. Normalisation de l'état
# ------------------------------------------------------------

X_normalise = (
    X_collectif - moyenne_etat
) / ecart_type_etat

print("\nÉtat normalisé construit.")
print("Forme de X_normalise :", X_normalise.shape)


# ------------------------------------------------------------
# 4. Calcul de la transition normalisée
# ------------------------------------------------------------

Delta_X_normalise = np.diff(
    X_normalise,
    axis=0
)

print(
    "Forme de Delta_X_normalise :",
    Delta_X_normalise.shape
)


# ------------------------------------------------------------
# 5. Vérification de la définition mathématique
# ------------------------------------------------------------

Delta_X_normalise_direct = (
    X_collectif[1:] - X_collectif[:-1]
) / ecart_type_etat

erreur_equivalence = np.max(
    np.abs(
        Delta_X_normalise
        -
        Delta_X_normalise_direct
    )
)

equivalence_validee = np.allclose(
    Delta_X_normalise,
    Delta_X_normalise_direct,
    rtol=1e-12,
    atol=1e-12
)

print("\nVérification :")
print(
    "Delta X_tilde = "
    "N(X_{k+1}) - N(X_k) :",
    equivalence_validee
)

print(
    "Erreur maximale d'équivalence :",
    f"{erreur_equivalence:.3e}"
)


# ------------------------------------------------------------
# 6. Contrôles statistiques de la normalisation
# ------------------------------------------------------------

moyennes_normalisees = np.mean(
    X_normalise,
    axis=0
)

ecarts_types_normalises = np.std(
    X_normalise,
    axis=0,
    ddof=0
)

erreur_moyenne_max = np.max(
    np.abs(moyennes_normalisees)
)

erreur_ecart_type_max = np.max(
    np.abs(ecarts_types_normalises - 1.0)
)

print("\nContrôles de normalisation :")
print(
    "- Moyenne absolue maximale :",
    f"{erreur_moyenne_max:.3e}"
)
print(
    "- Écart maximal à sigma = 1 :",
    f"{erreur_ecart_type_max:.3e}"
)
print(
    "- Absence de NaN :",
    not np.isnan(X_normalise).any()
)
print(
    "- Valeurs finies :",
    np.isfinite(X_normalise).all()
)


# ------------------------------------------------------------
# 7. Tableau des paramètres de normalisation
# ------------------------------------------------------------

parametres_normalisation_df = pd.DataFrame({
    "Composante": colonnes_etat,
    "Moyenne": moyenne_etat,
    "Ecart_type": ecart_type_etat
})

print("\nParamètres de normalisation :")
display(parametres_normalisation_df)


# ------------------------------------------------------------
# 8. DataFrame des transitions normalisées
# ------------------------------------------------------------

colonnes_transition_normalisee = [
    f"Delta_tilde_{colonne}"
    for colonne in colonnes_etat
]

Delta_X_normalise_df = pd.DataFrame(
    Delta_X_normalise,
    columns=colonnes_transition_normalisee
)

Delta_X_normalise_df.insert(
    0,
    "Calendar_Date_TDB",
    dates_transition
)

Delta_X_normalise_df.insert(
    0,
    "JDTDB",
    temps_transition
)

print("\nPremières transitions normalisées :")
display(Delta_X_normalise_df.head(3))
```

```python
# ============================================================
# Première observable fondamentale GVH
# D_T(k) = || Delta X_tilde_k ||_2
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


# ------------------------------------------------------------
# 1. Calcul de la norme euclidienne collective
# ------------------------------------------------------------

D_T = np.linalg.norm(
    Delta_X_normalise,
    axis=1,
    ord=2
)

print("Observable D_T calculée.")
print("Forme de D_T :", D_T.shape)


# ------------------------------------------------------------
# 2. Contrôles fondamentaux
# ------------------------------------------------------------

assert D_T.shape == (8760,)
assert np.isfinite(D_T).all()
assert not np.isnan(D_T).any()
assert np.all(D_T >= 0)

print("\nContrôles de D_T :")
print("- Nombre de valeurs : VALIDÉ")
print("- Absence de NaN : VALIDÉ")
print("- Valeurs finies : VALIDÉ")
print("- Positivité : VALIDÉ")


# ------------------------------------------------------------
# 3. Vérification directe pour la première transition
# ------------------------------------------------------------

D_T_0_direct = np.sqrt(
    np.sum(
        Delta_X_normalise[0] ** 2
    )
)

verification_D_T = np.isclose(
    D_T[0],
    D_T_0_direct,
    rtol=1e-14,
    atol=1e-14
)

print("\nVérification directe :")
print(
    "D_T(0) = sqrt(sum_j Delta_tilde_X(0,j)^2) :",
    verification_D_T
)
print("D_T(0) :", D_T[0])


# ------------------------------------------------------------
# 4. Distance métrique entre états normalisés
# ------------------------------------------------------------

D_T_metrique = np.linalg.norm(
    X_normalise[1:] - X_normalise[:-1],
    axis=1
)

erreur_metrique = np.max(
    np.abs(
        D_T - D_T_metrique
    )
)

compatibilite_metrique = np.allclose(
    D_T,
    D_T_metrique,
    rtol=1e-13,
    atol=1e-13
)

print("\nCompatibilité métrique :")
print(
    "D_T(k) = d(X_tilde_k, X_tilde_{k+1}) :",
    compatibilite_metrique
)
print(
    "Erreur maximale :",
    f"{erreur_metrique:.3e}"
)


# ------------------------------------------------------------
# 5. Statistiques descriptives
# ------------------------------------------------------------

resume_D_T = pd.DataFrame({
    "Observable": [
        "N_transitions",
        "D_T_min",
        "D_T_mean",
        "D_T_median",
        "D_T_std",
        "D_T_max",
        "Q05",
        "Q25",
        "Q75",
        "Q95"
    ],
    "Valeur": [
        len(D_T),
        np.min(D_T),
        np.mean(D_T),
        np.median(D_T),
        np.std(D_T, ddof=0),
        np.max(D_T),
        np.quantile(D_T, 0.05),
        np.quantile(D_T, 0.25),
        np.quantile(D_T, 0.75),
        np.quantile(D_T, 0.95)
    ]
})

print("\nRésumé statistique de D_T :")
display(resume_D_T)


# ------------------------------------------------------------
# 6. Contribution de chaque satellite
# ------------------------------------------------------------

Delta_Io = Delta_X_normalise[:, 0:6]
Delta_Europa = Delta_X_normalise[:, 6:12]
Delta_Ganymede = Delta_X_normalise[:, 12:18]

D_T_Io = np.linalg.norm(
    Delta_Io,
    axis=1
)

D_T_Europa = np.linalg.norm(
    Delta_Europa,
    axis=1
)

D_T_Ganymede = np.linalg.norm(
    Delta_Ganymede,
    axis=1
)


# Vérification de la décomposition quadratique
D_T_reconstruit = np.sqrt(
    D_T_Io**2
    + D_T_Europa**2
    + D_T_Ganymede**2
)

erreur_decomposition = np.max(
    np.abs(
        D_T - D_T_reconstruit
    )
)

print("\nDécomposition par satellite :")
print(
    "D_T² = D_T_Io² + D_T_Europa² + D_T_Ganymede² :",
    np.allclose(
        D_T,
        D_T_reconstruit,
        rtol=1e-13,
        atol=1e-13
    )
)
print(
    "Erreur maximale :",
    f"{erreur_decomposition:.3e}"
)


resume_satellites = pd.DataFrame({
    "Satellite": [
        "Io",
        "Europa",
        "Ganymede"
    ],
    "D_T_mean": [
        np.mean(D_T_Io),
        np.mean(D_T_Europa),
        np.mean(D_T_Ganymede)
    ],
    "D_T_std": [
        np.std(D_T_Io, ddof=0),
        np.std(D_T_Europa, ddof=0),
        np.std(D_T_Ganymede, ddof=0)
    ],
    "D_T_min": [
        np.min(D_T_Io),
        np.min(D_T_Europa),
        np.min(D_T_Ganymede)
    ],
    "D_T_max": [
        np.max(D_T_Io),
        np.max(D_T_Europa),
        np.max(D_T_Ganymede)
    ]
})

print("\nRésumé par satellite :")
display(resume_satellites)


# ------------------------------------------------------------
# 7. DataFrame final de l'observable
# ------------------------------------------------------------

D_T_df = pd.DataFrame({
    "JDTDB": temps_transition,
    "Calendar_Date_TDB": dates_transition,
    "D_T": D_T,
    "D_T_Io": D_T_Io,
    "D_T_Europa": D_T_Europa,
    "D_T_Ganymede": D_T_Ganymede
})

print("\nPremières valeurs de D_T :")
display(D_T_df.head(5))


# ------------------------------------------------------------
# 8. Visualisation temporelle
# ------------------------------------------------------------

plt.figure(figsize=(14, 5))

plt.plot(
    D_T_df["JDTDB"],
    D_T_df["D_T"],
    linewidth=0.8
)

plt.xlabel("JDTDB")
plt.ylabel(r"$D_T(k)$")
plt.title(
    "GVH — Norme de la transition normalisée\n"
    "Io, Europa et Ganymede"
)

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()


# ------------------------------------------------------------
# 9. Histogramme
# ------------------------------------------------------------

plt.figure(figsize=(9, 5))

plt.hist(
    D_T,
    bins=60,
    edgecolor="black",
    alpha=0.8
)

plt.xlabel(r"$D_T$")
plt.ylabel("Fréquence")
plt.title(
    "Distribution de la norme de transition normalisée"
)

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

### Résultat numérique fondamental

L'application aux trajectoires jovicentriques de Io, Europa et
Ganymède permet de construire un état collectif

\[
X_k \in \mathbb{R}^{18},
\]

composé des positions et vitesses des trois satellites.

Après normalisation composante par composante, la transition dynamique
est définie par

\[
\Delta\widetilde X_k
=
\widetilde X_{k+1}
-
\widetilde X_k.
\]

La première observable fondamentale de GVH est alors

\[
D_T(k)
=
\left\|
\Delta\widetilde X_k
\right\|_2.
\]

Le calcul numérique sur 8760 transitions horaires donne

\[
\overline{D_T}=0.414467,
\qquad
\sigma(D_T)=0.001613,
\]

avec

\[
0.412056
\leq D_T(k)
\leq
0.417890.
\]

L'identité métrique

\[
D_T(k)
=
d(
\widetilde X_k,
\widetilde X_{k+1}
)
\]

est vérifiée numériquement avec une erreur maximale nulle à la
précision de calcul.

La décomposition par satellite satisfait également

\[
D_T(k)^2
=
D_{T,\mathrm{Io}}(k)^2
+
D_{T,\mathrm{Europa}}(k)^2
+
D_{T,\mathrm{Ganymede}}(k)^2,
\]

avec une erreur maximale de \(1.665\times10^{-16}\).

Cette application établit que \(D_T\) n'est pas seulement une distance
abstraite : elle représente la norme de la transition dynamique
observée entre deux états successifs du système normalisé.

```python
# ============================================================
# Formalisation de S_T et S_T_norm
# Système collectif + contributions individuelles
# ============================================================

import numpy as np
import pandas as pd


# ------------------------------------------------------------
# 1. Fonction de calcul
# ------------------------------------------------------------

def calculer_variabilite_transition(
    distances,
    nom_systeme
):
    """
    Calcule les observables de variabilité de la norme de transition.

    Paramètres
    ----------
    distances : array-like
        Série temporelle D_T(k).
    nom_systeme : str
        Nom du système ou du satellite.

    Retour
    ------
    dict
        Résumé contenant :
        - nombre de transitions ;
        - moyenne de D_T ;
        - S_T ;
        - S_T_norm ;
        - variance ;
        - minimum et maximum.
    """

    distances = np.asarray(
        distances,
        dtype=float
    )

    # Contrôles d'intégrité
    if distances.ndim != 1:
        raise ValueError(
            f"{nom_systeme} : D_T doit être un vecteur unidimensionnel."
        )

    if distances.size == 0:
        raise ValueError(
            f"{nom_systeme} : la série D_T est vide."
        )

    if not np.isfinite(distances).all():
        raise ValueError(
            f"{nom_systeme} : D_T contient des valeurs non finies."
        )

    if np.any(distances < 0):
        raise ValueError(
            f"{nom_systeme} : D_T contient des valeurs négatives."
        )

    # Moyenne de la norme de transition
    D_T_mean = np.mean(
        distances
    )

    # Écart-type population
    S_T = np.std(
        distances,
        ddof=0
    )

    # Variance correspondante
    variance_D_T = np.var(
        distances,
        ddof=0
    )

    # Variabilité relative
    if np.isclose(D_T_mean, 0.0):
        S_T_norm = np.nan
    else:
        S_T_norm = S_T / D_T_mean

    return {
        "Systeme": nom_systeme,
        "N_transitions": distances.size,
        "D_T_mean": D_T_mean,
        "S_T": S_T,
        "S_T_norm": S_T_norm,
        "S_T_norm_percent": 100.0 * S_T_norm,
        "Variance_D_T": variance_D_T,
        "D_T_min": np.min(distances),
        "D_T_max": np.max(distances)
    }


# ------------------------------------------------------------
# 2. Calcul pour le système collectif
# ------------------------------------------------------------

resultat_collectif = calculer_variabilite_transition(
    D_T,
    "Io-Europa-Ganymede collectif"
)


# ------------------------------------------------------------
# 3. Calcul pour chaque satellite
# ------------------------------------------------------------

resultat_Io = calculer_variabilite_transition(
    D_T_Io,
    "Io"
)

resultat_Europa = calculer_variabilite_transition(
    D_T_Europa,
    "Europa"
)

resultat_Ganymede = calculer_variabilite_transition(
    D_T_Ganymede,
    "Ganymede"
)


# ------------------------------------------------------------
# 4. Tableau récapitulatif
# ------------------------------------------------------------

resume_S_T_df = pd.DataFrame([
    resultat_collectif,
    resultat_Io,
    resultat_Europa,
    resultat_Ganymede
])

print("Observables S_T et S_T_norm calculées.")
display(resume_S_T_df)


# ------------------------------------------------------------
# 5. Extraction explicite des observables collectives
# ------------------------------------------------------------

S_T = resultat_collectif["S_T"]
S_T_norm = resultat_collectif["S_T_norm"]

S_T_Io = resultat_Io["S_T"]
S_T_norm_Io = resultat_Io["S_T_norm"]

S_T_Europa = resultat_Europa["S_T"]
S_T_norm_Europa = resultat_Europa["S_T_norm"]

S_T_Ganymede = resultat_Ganymede["S_T"]
S_T_norm_Ganymede = resultat_Ganymede["S_T_norm"]


# ------------------------------------------------------------
# 6. Vérification directe de la définition de S_T
# ------------------------------------------------------------

S_T_direct = np.sqrt(
    np.mean(
        (D_T - np.mean(D_T)) ** 2
    )
)

verification_S_T = np.isclose(
    S_T,
    S_T_direct,
    rtol=1e-14,
    atol=1e-14
)

print("\nVérification mathématique :")
print(
    "S_T = sqrt(mean((D_T - mean(D_T))²)) :",
    verification_S_T
)

print(
    "Erreur absolue :",
    f"{abs(S_T - S_T_direct):.3e}"
)


# ------------------------------------------------------------
# 7. Vérification de la version normalisée
# ------------------------------------------------------------

S_T_norm_direct = (
    np.std(D_T, ddof=0)
    /
    np.mean(D_T)
)

verification_S_T_norm = np.isclose(
    S_T_norm,
    S_T_norm_direct,
    rtol=1e-14,
    atol=1e-14
)

print(
    "S_T_norm = S_T / mean(D_T) :",
    verification_S_T_norm
)

print(
    "Erreur absolue :",
    f"{abs(S_T_norm - S_T_norm_direct):.3e}"
)


# ------------------------------------------------------------
# 8. Contrôles globaux
# ------------------------------------------------------------

observables_calculees = np.array([
    S_T,
    S_T_norm,
    S_T_Io,
    S_T_norm_Io,
    S_T_Europa,
    S_T_norm_Europa,
    S_T_Ganymede,
    S_T_norm_Ganymede
])

print("\nContrôles des observables :")
print(
    "- Valeurs finies :",
    np.isfinite(observables_calculees).all()
)
print(
    "- S_T non négatifs :",
    np.all(
        np.array([
            S_T,
            S_T_Io,
            S_T_Europa,
            S_T_Ganymede
        ]) >= 0
    )
)
print(
    "- S_T_norm non négatifs :",
    np.all(
        np.array([
            S_T_norm,
            S_T_norm_Io,
            S_T_norm_Europa,
            S_T_norm_Ganymede
        ]) >= 0
    )
)


# ------------------------------------------------------------
# 9. Affichage scientifique compact
# ------------------------------------------------------------

print("\nRésultats collectifs :")
print(
    f"D_T_mean  = {np.mean(D_T):.9f}"
)
print(
    f"S_T       = {S_T:.9f}"
)
print(
    f"S_T_norm  = {S_T_norm:.9f}"
)
print(
    f"S_T_norm  = {100 * S_T_norm:.6f} %"
)

print("\nRésultats individuels :")

print(
    f"Io        : "
    f"S_T = {S_T_Io:.9f}, "
    f"S_T_norm = {S_T_norm_Io:.9f} "
    f"({100 * S_T_norm_Io:.6f} %)"
)

print(
    f"Europa    : "
    f"S_T = {S_T_Europa:.9f}, "
    f"S_T_norm = {S_T_norm_Europa:.9f} "
    f"({100 * S_T_norm_Europa:.6f} %)"
)

print(
    f"Ganymede  : "
    f"S_T = {S_T_Ganymede:.9f}, "
    f"S_T_norm = {S_T_norm_Ganymede:.9f} "
    f"({100 * S_T_norm_Ganymede:.6f} %)"
)
```
