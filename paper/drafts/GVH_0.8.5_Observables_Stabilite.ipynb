```python
# ============================================================
# Construction des transitions — Europa 2026
# ============================================================

import numpy as np
import pandas as pd

def construire_transitions(df, satellite):
    colonnes_etat = ["X", "Y", "Z", "VX", "VY", "VZ"]

    variations = (
        df[colonnes_etat]
        .diff()
        .iloc[1:]
        .reset_index(drop=True)
    )

    variations.columns = [
        "dX", "dY", "dZ",
        "dVX", "dVY", "dVZ"
    ]

    transitions = pd.DataFrame({
        "Satellite": satellite,
        "Datetime": df["Datetime"].iloc[1:].reset_index(drop=True),
        "dX": variations["dX"],
        "dY": variations["dY"],
        "dZ": variations["dZ"],
        "dVX": variations["dVX"],
        "dVY": variations["dVY"],
        "dVZ": variations["dVZ"],
    })

    transitions["D_T"] = np.sqrt(
        transitions["dX"]**2
        + transitions["dY"]**2
        + transitions["dZ"]**2
    )

    return transitions


transitions_europa_2026 = construire_transitions(
    europa_2026,
    "Europa"
)

print("Dimensions :", transitions_europa_2026.shape)
print("Valeurs manquantes :", transitions_europa_2026.isna().sum().sum())
print("D_T moyen :", transitions_europa_2026["D_T"].mean())
print("D_T min :", transitions_europa_2026["D_T"].min())
print("D_T max :", transitions_europa_2026["D_T"].max())

display(transitions_europa_2026.head())
```

```python
# ============================================================
# Observables GVH de base — Europa 2026
# ============================================================

import numpy as np
import pandas as pd

def calculer_observables_gvh_base(
    transitions,
    satellite,
    quantile_seuil=0.95
):
    D_T = transitions["D_T"].to_numpy(dtype=float)

    # Observable moyenne
    D_T_mean = np.mean(D_T)

    # Dispersion absolue
    S_T = np.std(D_T, ddof=0)

    # Dispersion normalisée
    S_T_norm = S_T / D_T_mean if D_T_mean != 0 else np.nan

    # Normalisation dans [0, 1]
    D_T_min = np.min(D_T)
    D_T_max = np.max(D_T)

    amplitude = D_T_max - D_T_min

    if amplitude > 0:
        u = (D_T - D_T_min) / amplitude
    else:
        u = np.zeros_like(D_T)

    # Seuil critique
    u_c = np.quantile(u, quantile_seuil)

    # D_c exprimé dans l'unité physique de D_T
    D_c = D_T_min + u_c * amplitude

    # Fraction située au-dessus du seuil
    R_super = np.mean(u > u_c)

    observables = pd.DataFrame([{
        "Satellite": satellite,
        "Annee": 2026,
        "N_etats": len(transitions) + 1,
        "N_transitions": len(transitions),
        "D_T_mean": D_T_mean,
        "S_T": S_T,
        "S_T_norm": S_T_norm,
        "D_T_min": D_T_min,
        "D_T_max": D_T_max,
        "D_c": D_c,
        "u_c": u_c,
        "R_super": R_super
    }])

    transitions_gvh = transitions.copy()
    transitions_gvh["u"] = u
    transitions_gvh["supercritique"] = u > u_c

    return observables, transitions_gvh


observables_europa_2026_base, transitions_europa_2026_gvh = (
    calculer_observables_gvh_base(
        transitions_europa_2026,
        satellite="Europa",
        quantile_seuil=0.95
    )
)

display(observables_europa_2026_base)

print("\nContrôles :")
print("u min :", transitions_europa_2026_gvh["u"].min())
print("u max :", transitions_europa_2026_gvh["u"].max())
print(
    "Nombre de transitions supercritiques :",
    transitions_europa_2026_gvh["supercritique"].sum()
)
```

```python
# ============================================================
# Observables angulaires — Europa 2026
# ============================================================

import numpy as np
import pandas as pd

def calculer_observables_angulaires(transitions, satellite, annee):
    vecteurs = transitions[["dX", "dY", "dZ"]].to_numpy(dtype=float)

    # Comparaison entre deux vecteurs de transition successifs
    v1 = vecteurs[:-1]
    v2 = vecteurs[1:]

    normes_v1 = np.linalg.norm(v1, axis=1)
    normes_v2 = np.linalg.norm(v2, axis=1)

    denominateur = normes_v1 * normes_v2

    # Évite toute division par zéro
    masque_valide = denominateur > 0

    cos_theta = np.full(len(denominateur), np.nan)

    cos_theta[masque_valide] = (
        np.sum(v1[masque_valide] * v2[masque_valide], axis=1)
        / denominateur[masque_valide]
    )

    # Correction des erreurs numériques éventuelles
    cos_theta = np.clip(cos_theta, -1.0, 1.0)

    theta_deg = np.degrees(np.arccos(cos_theta))
    theta_valide = theta_deg[np.isfinite(theta_deg)]

    if len(theta_valide) == 0:
        raise ValueError("Aucun angle valide n'a pu être calculé.")

    angles = pd.DataFrame({
        "Satellite": satellite,
        "Datetime": transitions["Datetime"].iloc[1:].reset_index(drop=True),
        "theta_deg": theta_deg
    })

    observables = pd.DataFrame([{
        "Satellite": satellite,
        "Annee": annee,
        "N_angles": len(theta_valide),
        "theta_mean": np.mean(theta_valide),
        "theta_median": np.median(theta_valide),
        "theta_std": np.std(theta_valide, ddof=0),
        "theta_min": np.min(theta_valide),
        "theta_max": np.max(theta_valide),
        "R_90": np.mean(theta_valide >= 90),
        "R_170": np.mean(theta_valide >= 170),
        "R_175": np.mean(theta_valide >= 175),
        "R_180": np.mean(theta_valide >= 179.9)
    }])

    return observables, angles


observables_europa_2026_angles, angles_europa_2026 = (
    calculer_observables_angulaires(
        transitions_europa_2026,
        satellite="Europa",
        annee=2026
    )
)

display(observables_europa_2026_angles)

print("\nContrôles :")
print("Nombre attendu d'angles :", len(transitions_europa_2026) - 1)
print("Nombre calculé :", angles_europa_2026["theta_deg"].notna().sum())
print("Valeurs manquantes :", angles_europa_2026["theta_deg"].isna().sum())

display(angles_europa_2026.head())
```

```python
# ============================================================
# Fusion des observables Europa 2026
# ============================================================

observables_europa_2026 = (
    observables_europa_2026_base
    .merge(
        observables_europa_2026_angles,
        on=["Satellite", "Annee"],
        how="left"
    )
)

display(observables_europa_2026)
```
