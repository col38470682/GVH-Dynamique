```python
# ============================================================
# Nettoyage complet des fichiers NASA/JPL Horizons — 2013
# Io • Europa • Ganymède
# ============================================================

import pandas as pd

colonnes_horizons = [
    "JD",
    "Datetime",
    "X",
    "Y",
    "Z",
    "VX",
    "VY",
    "VZ",
    "LT",
    "RG",
    "RR"
]

def nettoyer_horizons(df_brut):
    """
    Nettoie un DataFrame brut issu de NASA/JPL Horizons.

    - Supprime la colonne vide finale éventuelle
    - Nomme les colonnes
    - Convertit les colonnes numériques
    - Convertit correctement les dates Horizons
    """

    df = df_brut.copy()

    # Vérification du nombre de colonnes
    if df.shape[1] == 12:
        # La 12e colonne est vide à cause de la virgule finale Horizons
        df = df.iloc[:, :11]

    if df.shape[1] != 11:
        raise ValueError(
            f"Nombre de colonnes inattendu : {df.shape[1]}. "
            "Le fichier devrait contenir 11 colonnes utiles."
        )

    # Attribution des noms de colonnes
    df.columns = colonnes_horizons

    # Conversion des colonnes numériques
    colonnes_numeriques = [
        "JD",
        "X",
        "Y",
        "Z",
        "VX",
        "VY",
        "VZ",
        "LT",
        "RG",
        "RR"
    ]

    for colonne in colonnes_numeriques:
        df[colonne] = pd.to_numeric(
            df[colonne],
            errors="coerce"
        )

    # Nettoyage du format Horizons :
    # A.D. 2013-Jan-01 00:00:00.0000
    datetime_nettoye = (
        df["Datetime"]
        .astype(str)
        .str.strip()
        .str.replace("A.D. ", "", regex=False)
    )

    df["Datetime"] = pd.to_datetime(
        datetime_nettoye,
        format="%Y-%b-%d %H:%M:%S.%f",
        errors="coerce"
    )

    # Réinitialisation propre de l'index
    df = df.reset_index(drop=True)

    return df


# ============================================================
# Application aux trois satellites
# ============================================================

io_2013 = nettoyer_horizons(io_2013_brut)
europa_2013 = nettoyer_horizons(europa_2013_brut)
ganymede_2013 = nettoyer_horizons(ganymede_2013_brut)


# ============================================================
# Validation rapide
# ============================================================

def validation_rapide(nom, df):
    print(f"\n--- {nom} ---")
    print("Dimensions :", df.shape)
    print("Début :", df["Datetime"].min())
    print("Fin :", df["Datetime"].max())
    print("Valeurs manquantes totales :", df.isna().sum().sum())
    print("Dates invalides :", df["Datetime"].isna().sum())
    print("Doublons temporels :", df["Datetime"].duplicated().sum())

    cadence = df["Datetime"].sort_values().diff().dropna()

    if cadence.empty:
        print("Cadence : impossible à calculer")
    else:
        mode_cadence = cadence.mode()

        if not mode_cadence.empty:
            print("Cadence la plus fréquente :", mode_cadence.iloc[0])

        print("Cadence minimale :", cadence.min())
        print("Cadence maximale :", cadence.max())


validation_rapide("Io", io_2013)
validation_rapide("Europa", europa_2013)
validation_rapide("Ganymede", ganymede_2013)


# ============================================================
# Aperçu final
# ============================================================

display(io_2013.head())
```

```python
# ============================================================
# Construction des transitions physiques — 2013
# ============================================================

import numpy as np
import pandas as pd

colonnes_etat = ["X", "Y", "Z", "VX", "VY", "VZ"]

def construire_transitions(nom, df):
    """
    Construit les transitions horaires dans l'espace d'état 6D.
    """

    etats = df[colonnes_etat].to_numpy(dtype=float)

    # Différences entre deux états horaires successifs
    delta = np.diff(etats, axis=0)

    # Norme euclidienne 6D
    D_T = np.linalg.norm(delta, axis=1)

    transitions = pd.DataFrame({
        "Satellite": nom,
        "Datetime_debut": df["Datetime"].iloc[:-1].to_numpy(),
        "Datetime_fin": df["Datetime"].iloc[1:].to_numpy(),
        "Delta_X": delta[:, 0],
        "Delta_Y": delta[:, 1],
        "Delta_Z": delta[:, 2],
        "Delta_VX": delta[:, 3],
        "Delta_VY": delta[:, 4],
        "Delta_VZ": delta[:, 5],
        "D_T": D_T
    })

    return transitions


transitions_io_2013 = construire_transitions("Io", io_2013)
transitions_europa_2013 = construire_transitions("Europa", europa_2013)
transitions_ganymede_2013 = construire_transitions(
    "Ganymede",
    ganymede_2013
)

print("Io :", transitions_io_2013.shape)
print("Europa :", transitions_europa_2013.shape)
print("Ganymede :", transitions_ganymede_2013.shape)

display(transitions_io_2013.head())
```

```python
# ============================================================
# Statistiques physiques initiales de D_T
# ============================================================

def statistiques_DT(nom, transitions):
    D_T = transitions["D_T"]

    return {
        "Satellite": nom,
        "N_transitions": len(D_T),
        "D_T_mean": D_T.mean(),
        "S_T": D_T.std(ddof=0),
        "D_T_min": D_T.min(),
        "D_T_max": D_T.max()
    }


statistiques_physiques_2013 = pd.DataFrame([
    statistiques_DT("Io", transitions_io_2013),
    statistiques_DT("Europa", transitions_europa_2013),
    statistiques_DT("Ganymede", transitions_ganymede_2013)
])

display(statistiques_physiques_2013)
```

```python
# ============================================================
# Observables GVH normalisées — 2013
# Définitions figées identiques à 2000 et 2026
# ============================================================

QUANTILE_GVH = 0.95

def calculer_observables_gvh(nom, transitions, quantile=QUANTILE_GVH):
    D_T = transitions["D_T"].to_numpy(dtype=float)

    D_T_mean = np.mean(D_T)
    S_T = np.std(D_T, ddof=0)
    D_T_max = np.max(D_T)

    # Normalisation figée
    u = D_T / D_T_max

    S_T_norm = S_T / D_T_max

    # Seuil au quantile 95 %
    D_c = np.quantile(D_T, quantile)
    u_c = np.quantile(u, quantile)

    # Fraction strictement supérieure au seuil
    R_super = np.mean(u > u_c)

    # Ajout de u dans la table des transitions
    transitions_gvh = transitions.copy()
    transitions_gvh["u"] = u
    transitions_gvh["Supercritique"] = u > u_c

    observables = {
        "Satellite": nom,
        "N_etats": len(D_T) + 1,
        "N_transitions": len(D_T),
        "Quantile": quantile,
        "D_T_mean": D_T_mean,
        "S_T": S_T,
        "S_T_norm": S_T_norm,
        "D_T_max": D_T_max,
        "D_c": D_c,
        "u_c": u_c,
        "R_super": R_super
    }

    return observables, transitions_gvh


obs_io_2013, transitions_io_2013 = calculer_observables_gvh(
    "Io",
    transitions_io_2013
)

obs_europa_2013, transitions_europa_2013 = calculer_observables_gvh(
    "Europa",
    transitions_europa_2013
)

obs_ganymede_2013, transitions_ganymede_2013 = calculer_observables_gvh(
    "Ganymede",
    transitions_ganymede_2013
)

observables_gvh_2013_base = pd.DataFrame([
    obs_io_2013,
    obs_europa_2013,
    obs_ganymede_2013
])

display(observables_gvh_2013_base)
```

```python
# ============================================================
# Observables angulaires GVH — 2013
# Angle entre deux transitions successives dans l'espace 6D
# ============================================================

import numpy as np
import pandas as pd

def calculer_observables_angulaires(nom, transitions):
    """
    Calcule l'angle entre deux vecteurs de transition successifs.

    Le vecteur utilisé est :
    (Delta_X, Delta_Y, Delta_Z, Delta_VX, Delta_VY, Delta_VZ)
    """

    colonnes_delta = [
        "Delta_X",
        "Delta_Y",
        "Delta_Z",
        "Delta_VX",
        "Delta_VY",
        "Delta_VZ"
    ]

    vecteurs = transitions[colonnes_delta].to_numpy(dtype=float)

    vecteurs_1 = vecteurs[:-1]
    vecteurs_2 = vecteurs[1:]

    produits_scalaires = np.sum(vecteurs_1 * vecteurs_2, axis=1)

    normes_1 = np.linalg.norm(vecteurs_1, axis=1)
    normes_2 = np.linalg.norm(vecteurs_2, axis=1)

    denominateur = normes_1 * normes_2

    masque_valide = denominateur > 0

    cos_theta = np.full(len(denominateur), np.nan)

    cos_theta[masque_valide] = (
        produits_scalaires[masque_valide]
        / denominateur[masque_valide]
    )

    # Protection contre les erreurs d'arrondi numérique
    cos_theta = np.clip(cos_theta, -1.0, 1.0)

    theta = np.degrees(np.arccos(cos_theta))

    theta_valide = theta[np.isfinite(theta)]

    observables = {
        "Satellite": nom,
        "N_angles": len(theta_valide),
        "theta_mean": np.mean(theta_valide),
        "theta_median": np.median(theta_valide),
        "theta_std": np.std(theta_valide, ddof=0),
        "theta_min": np.min(theta_valide),
        "theta_max": np.max(theta_valide),
        "R_170": np.mean(theta_valide >= 170.0),
        "R_175": np.mean(theta_valide >= 175.0),
        "R_180": np.mean(theta_valide >= 179.0)
    }

    table_angles = pd.DataFrame({
        "Satellite": nom,
        "Datetime_centre": transitions["Datetime_fin"].iloc[:-1].to_numpy(),
        "theta_deg": theta
    })

    return observables, table_angles


angles_obs_io_2013, angles_io_2013 = calculer_observables_angulaires(
    "Io",
    transitions_io_2013
)

angles_obs_europa_2013, angles_europa_2013 = calculer_observables_angulaires(
    "Europa",
    transitions_europa_2013
)

angles_obs_ganymede_2013, angles_ganymede_2013 = calculer_observables_angulaires(
    "Ganymede",
    transitions_ganymede_2013
)

observables_angulaires_2013 = pd.DataFrame([
    angles_obs_io_2013,
    angles_obs_europa_2013,
    angles_obs_ganymede_2013
])

display(observables_angulaires_2013)
```

```python
# ============================================================
# Estimation des périodes orbitales — 2013
# Méthode : angle orbital déroulé dans le plan XY
# ============================================================

import numpy as np
import pandas as pd

def estimer_periode_orbitale(nom, df):
    """
    Estime la période orbitale moyenne à partir de l'angle atan2(Y, X).
    """

    temps_heures = (
        df["Datetime"] - df["Datetime"].iloc[0]
    ).dt.total_seconds().to_numpy() / 3600.0

    angle_rad = np.arctan2(
        df["Y"].to_numpy(dtype=float),
        df["X"].to_numpy(dtype=float)
    )

    # Déroulement continu de l'angle
    angle_unwrap = np.unwrap(angle_rad)

    # Régression linéaire : angle = omega * temps + constante
    coefficient = np.polyfit(
        temps_heures,
        angle_unwrap,
        deg=1
    )

    omega_rad_h = coefficient[0]

    periode_heures = 2.0 * np.pi / abs(omega_rad_h)
    periode_jours = periode_heures / 24.0

    # Reconstruction linéaire et résidus
    angle_modele = np.polyval(coefficient, temps_heures)
    residus = angle_unwrap - angle_modele

    ss_res = np.sum(residus**2)
    ss_tot = np.sum(
        (angle_unwrap - np.mean(angle_unwrap))**2
    )

    r2 = 1.0 - ss_res / ss_tot if ss_tot > 0 else np.nan

    resultat = {
        "Satellite": nom,
        "omega_rad_h": omega_rad_h,
        "periode_heures": periode_heures,
        "periode_jours": periode_jours,
        "R2_angle_temps": r2,
        "residu_angle_std_rad": np.std(residus, ddof=0)
    }

    table_angle = pd.DataFrame({
        "Satellite": nom,
        "Datetime": df["Datetime"],
        "temps_heures": temps_heures,
        "angle_rad": angle_rad,
        "angle_unwrap_rad": angle_unwrap,
        "angle_modele_rad": angle_modele,
        "residu_angle_rad": residus
    })

    return resultat, table_angle


periode_io_2013, phase_io_2013 = estimer_periode_orbitale(
    "Io",
    io_2013
)

periode_europa_2013, phase_europa_2013 = estimer_periode_orbitale(
    "Europa",
    europa_2013
)

periode_ganymede_2013, phase_ganymede_2013 = estimer_periode_orbitale(
    "Ganymede",
    ganymede_2013
)

periodes_orbitales_2013 = pd.DataFrame([
    periode_io_2013,
    periode_europa_2013,
    periode_ganymede_2013
])

display(periodes_orbitales_2013)
```
