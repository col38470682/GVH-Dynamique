## Objectif

Ce notebook construit les fondations mathématiques du framework GVH.

GVH ne remplace pas les lois physiques fondamentales. Il utilise les
équations de la mécanique classique, de la gravitation et des systèmes
dynamiques afin de définir des observables capables de mesurer,
comparer et caractériser l’évolution d’un système dans son espace d’état.

Le développement suivra l’ordre suivant :

1. cadre physique de référence ;
2. définition de l’espace d’état ;
3. normalisation des variables ;
4. définition d’une métrique dynamique ;
5. définition des observables GVH ;
6. étude de leurs propriétés mathématiques ;
7. extension aux systèmes à N corps ;
8. préparation d’un module séparé consacré aux manœuvres orbitales.

## 1. Positionnement scientifique

Les lois de Newton déterminent l’évolution physique du mouvement.

Le framework GVH cherche à décrire la géométrie et les transitions de
cette évolution dans l’espace d’état.

La distinction fondamentale est donc :

\[
\text{Lois physiques}
\longrightarrow
\text{Évolution du système}
\longrightarrow
\text{Observables GVH}.
\]

Newton répond principalement à la question :

> Quelle force produit quelle accélération ?

GVH cherche à répondre à la question :

> Comment mesurer, comparer et classifier les changements dynamiques
> produits dans l’espace d’état ?

## 2. Première loi de Newton — Principe d’inertie

Dans un référentiel inertiel, tout corps conserve son état de repos ou
de mouvement rectiligne uniforme lorsque la résultante des forces
extérieures qui s’exercent sur lui est nulle.

Mathématiquement :

\[
\sum_i \mathbf{F}_i = \mathbf{0}
\quad \Longrightarrow \quad
\frac{d\mathbf{v}}{dt} = \mathbf{0}.
\]

La vitesse reste constante :

\[
\mathbf{v}(t)=\mathbf{v}_0.
\]

La position évolue alors selon :

\[
\mathbf{r}(t)=\mathbf{r}_0+\mathbf{v}_0t.
\]

Dans le framework GVH, cette évolution constitue un régime inertiel de
référence. Les observables doivent distinguer une évolution régulière
due au mouvement inertiel d’une véritable perturbation dynamique.

## 3. Deuxième loi de Newton — Principe fondamental de la dynamique

Dans un référentiel inertiel, la variation de la quantité de mouvement
d’un corps est égale à la résultante des forces extérieures qui
s’exercent sur lui.

La forme générale est :

\[
\frac{d\mathbf{p}}{dt}
=
\sum_i \mathbf{F}_i,
\]

où :

\[
\mathbf{p}=m\mathbf{v}
\]

est la quantité de mouvement.

Lorsque la masse \(m\) est constante :

\[
\sum_i \mathbf{F}_i
=
m\frac{d\mathbf{v}}{dt}
=
m\mathbf{a}.
\]

Ainsi :

\[
\mathbf{a}
=
\frac{1}{m}\sum_i \mathbf{F}_i.
\]

Cette loi relie directement :

\[
\text{forces}
\longrightarrow
\text{accélération}
\longrightarrow
\text{modification de la vitesse}
\longrightarrow
\text{modification de la trajectoire}.
\]

### 3.1 Formulation dans l’espace d’état

Pour un corps ponctuel, l’état dynamique est défini par :

\[
X(t)
=
\begin{pmatrix}
\mathbf{r}(t)\\
\mathbf{v}(t)
\end{pmatrix}
\in \mathbb{R}^{6},
\]

où :

- \(\mathbf{r}(t)\in\mathbb{R}^3\) est la position ;
- \(\mathbf{v}(t)\in\mathbb{R}^3\) est la vitesse.

L’évolution de cet état s’écrit :

\[
\frac{dX}{dt}
=
\begin{pmatrix}
\dot{\mathbf{r}}\\
\dot{\mathbf{v}}
\end{pmatrix}
=
\begin{pmatrix}
\mathbf{v}\\
\mathbf{a}
\end{pmatrix}.
\]

En utilisant la deuxième loi de Newton :

\[
\frac{dX}{dt}
=
\begin{pmatrix}
\mathbf{v}\\
\frac{1}{m}\sum_i\mathbf{F}_i
\end{pmatrix}.
\]

Nous pouvons donc définir le champ dynamique :

\[
\dot X(t)=F(X,t),
\]

avec :

\[
F(X,t)
=
\begin{pmatrix}
\mathbf{v}\\
\frac{1}{m}\sum_i\mathbf{F}_i(X,t)
\end{pmatrix}.
\]

Les lois physiques déterminent le champ dynamique \(F\), tandis que le
framework GVH analyse la trajectoire produite dans l’espace d’état.

### 3.2 Interprétation dans le framework GVH

La deuxième loi de Newton détermine comment une force modifie l’état
dynamique d’un système.

Entre deux instants successifs \(t_k\) et \(t_{k+1}\), la transition
d’état est :

\[
\Delta X_k
=
X(t_{k+1})-X(t_k).
\]

Cette transition contient deux composantes :

\[
\Delta X_k
=
\begin{pmatrix}
\Delta\mathbf{r}_k\\
\Delta\mathbf{v}_k
\end{pmatrix},
\]

avec :

\[
\Delta\mathbf{r}_k
=
\mathbf{r}(t_{k+1})-\mathbf{r}(t_k),
\]

et :

\[
\Delta\mathbf{v}_k
=
\mathbf{v}(t_{k+1})-\mathbf{v}(t_k).
\]

Pour un pas temporel suffisamment petit :

\[
\Delta\mathbf{v}_k
\approx
\mathbf{a}(t_k)\Delta t
=
\frac{\sum_i\mathbf{F}_i(t_k)}{m}\Delta t.
\]

Ainsi, les forces produisent une modification de la vitesse qui se
répercute ensuite sur la trajectoire.

Le framework GVH ne remplace pas l’équation :

\[
\sum_i\mathbf{F}_i=m\mathbf{a}.
\]

Il construit des observables à partir des transitions générées par cette
équation.

### 3.3 Transition totale et transition résiduelle

Une transition d’état non nulle ne signifie pas nécessairement qu’une
perturbation est présente.

Même dans un mouvement inertiel :

\[
\mathbf{r}(t+\Delta t)
=
\mathbf{r}(t)+\mathbf{v}(t)\Delta t,
\]

la position change naturellement.

Nous devons donc distinguer deux quantités.

#### Transition totale

\[
\Delta X_k^{\mathrm{tot}}
=
X(t_{k+1})-X(t_k).
\]

Elle mesure l’évolution complète du système.

#### État de référence inertiel

À partir de l’état observé à \(t_k\), l’état inertiel attendu au temps
\(t_{k+1}\) est :

\[
X_{k+1}^{\mathrm{inertiel}}
=
\begin{pmatrix}
\mathbf{r}_k+\mathbf{v}_k\Delta t\\
\mathbf{v}_k
\end{pmatrix}.
\]

#### Transition résiduelle

\[
\Delta X_k^{\mathrm{res}}
=
X(t_{k+1})
-
X_{k+1}^{\mathrm{inertiel}}.
\]

Dans un mouvement parfaitement inertiel :

\[
\Delta X_k^{\mathrm{res}}
=
\mathbf{0}.
\]

Lorsqu’une force agit :

\[
\Delta X_k^{\mathrm{res}}
\neq
\mathbf{0}.
\]

Cette quantité résiduelle permet au framework GVH de distinguer le
mouvement attendu de la modification dynamique produite par une
interaction.

### 3.4 Distance résiduelle GVH provisoire

Nous introduisons provisoirement l’observable :

\[
D_{\mathrm{res}}(k)
=
\left\|
X(t_{k+1})
-
X_{k+1}^{\mathrm{inertiel}}
\right\|.
\]

Elle mesure la distance entre :

- l’état réellement observé ;
- l’état qui aurait été obtenu par évolution inertielle.

Cependant, cette définition n’est pas encore définitive, car les
positions et les vitesses possèdent des unités différentes.

Une normalisation devra être introduite avant de définir rigoureusement
la métrique GVH.

La future expression prendra la forme :

\[
D_{\mathrm{res}}^{\mathrm{GVH}}(k)
=
d_{\mathcal X}
\left(
\widetilde X(t_{k+1}),
\widetilde X_{k+1}^{\mathrm{inertiel}}
\right),
\]

où :

- \(\widetilde X\) est un état normalisé ;
- \(d_{\mathcal X}\) est une métrique définie sur l’espace d’état.

## 4. Troisième loi de Newton — Principe d'action et de réaction

Lorsque deux corps interagissent, chacun exerce sur l'autre une force de
même intensité, de même direction, mais de sens opposé.

Mathématiquement,

\[
\mathbf{F}_{ij}
=
-\mathbf{F}_{ji}.
\]

Cette relation est valable pour toutes les interactions mutuelles
décrites par la mécanique classique.

Elle implique que les forces internes d'un système s'annulent deux à
deux lorsqu'elles sont sommées sur l'ensemble des corps.

### 4.1 Conservation de la quantité de mouvement

Considérons un système composé de \(N\) corps.

La quantité de mouvement totale est

\[
\mathbf P
=
\sum_{i=1}^{N}
m_i\mathbf v_i.
\]

En dérivant par rapport au temps,

\[
\frac{d\mathbf P}{dt}
=
\sum_{i=1}^{N}
m_i\mathbf a_i.
\]

En utilisant la deuxième loi de Newton,

\[
\frac{d\mathbf P}{dt}
=
\sum_{i=1}^{N}
\mathbf F_i.
\]

Lorsque seules des forces internes sont présentes,

\[
\sum_{i=1}^{N}
\mathbf F_i
=
0,
\]

car

\[
\mathbf F_{ij}
=
-
\mathbf F_{ji}.
\]

Par conséquent,

\[
\frac{d\mathbf P}{dt}
=
0.
\]

La quantité de mouvement totale demeure constante.

Cette propriété constitue l'un des principes fondamentaux de la mécanique classique.

### 4.2 Interprétation dans le framework GVH

Dans un système composé de plusieurs corps,

chaque interaction modifie simultanément plusieurs trajectoires.

Le framework GVH ne mesure pas directement les forces.

Il mesure les conséquences collectives de ces interactions sur
l'évolution de l'espace d'état.

Ainsi,

une interaction entre deux corps produit simultanément :

- une modification locale de leurs trajectoires ;

- une modification globale de la géométrie du système.

Les observables GVH seront construites pour quantifier cette évolution
collective.

## 5. Du problème à deux corps au problème N-corps

Les trois lois de Newton s'appliquent aussi bien à un corps qu'à un
ensemble de plusieurs corps.

Dans un système constitué de \(N\) objets,

chaque corps subit l'action gravitationnelle de tous les autres.

La force résultante appliquée au corps \(i\) est

\[
\mathbf F_i
=
\sum_{j\ne i}
\mathbf F_{ij}.
\]

Les équations du mouvement deviennent alors un système couplé où
l'évolution de chaque trajectoire dépend simultanément de toutes les
autres.

Le framework GVH est conçu pour analyser cette dynamique globale à partir
de l'évolution de l'espace d'état du système plutôt que d'étudier chaque
interaction séparément.

# 6. Espace d'état d'un système dynamique

Les lois de Newton déterminent l'évolution d'un système physique.

Afin d'analyser cette évolution de manière générale, le framework GVH
représente chaque système par un état appartenant à un espace d'état.

Un état contient toutes les informations nécessaires pour décrire
complètement le système à un instant donné.

Pour un corps ponctuel, l'état est défini par

$begin:math:display$
X\(t\)\=
\(\\mathbf r\(t\)\,\\mathbf v\(t\)\)\,
$end:math:display$

où

- $begin:math:text$\\mathbf r\(t\)$end:math:text$ est le vecteur position ;
- $begin:math:text$\\mathbf v\(t\)$end:math:text$ est le vecteur vitesse.

L'état appartient à

$begin:math:display$
\\mathcal X
\=
\\mathbb R\^3
\\times
\\mathbb R\^3
\=
\\mathbb R\^6\.
$end:math:display$

Chaque instant correspond donc à un point de l'espace d'état.

## 6.1 Espace d'état d'un système N-corps

Considérons un système composé de

$begin:math:display$
N
$end:math:display$

corps.

Pour chaque corps

$begin:math:display$
i
\=
1\,\\ldots\,N\,
$end:math:display$

on définit

$begin:math:display$
\\mathbf r\_i\(t\)
\\in
\\mathbb R\^3
$end:math:display$

et

$begin:math:display$
\\mathbf v\_i\(t\)
\\in
\\mathbb R\^3\.
$end:math:display$

L'état complet du système est

$begin:math:display$
X\(t\)
\=
\(
\\mathbf r\_1\,
\\mathbf v\_1\,
\\ldots\,
\\mathbf r\_N\,
\\mathbf v\_N
\)\.
$end:math:display$

Cet état appartient à

$begin:math:display$
\\mathcal X\_N
\=
\(\\mathbb R\^6\)\^N
\\simeq
\\mathbb R\^\{6N\}\.
$end:math:display$

Ainsi, un système à N corps est représenté par un unique point dans un
espace de dimension

$begin:math:display$
6N\.
$end:math:display$

## 6.2 Trajectoire dans l'espace d'état

Au cours du temps, le système évolue.

Cette évolution définit une courbe

$begin:math:display$
X \:
\\mathbb R
\\rightarrow
\\mathcal X\_N\.
$end:math:display$

Autrement dit,

à chaque instant

$begin:math:display$
t\,
$end:math:display$

le système occupe une position unique dans l'espace d'état.

Lorsque le temps évolue,

le point

$begin:math:display$
X\(t\)
$end:math:display$

décrit une trajectoire dans

$begin:math:display$
\\mathcal X\_N\.
$end:math:display$

Le framework GVH analysera cette trajectoire plutôt que les positions
physiques seules.

## 6.3 Vision du framework GVH

Dans la mécanique classique,

l'objectif principal est de déterminer les trajectoires physiques.

Dans le framework GVH,

l'objectif est de mesurer les propriétés géométriques de la trajectoire
dans l'espace d'état.

Les observables GVH ne remplacent donc pas les équations du mouvement.

Elles décrivent la manière dont un système évolue dans son espace d'état.

Cette approche permet d'appliquer un même formalisme à des systèmes
gravitationnels, chaotiques ou expérimentaux.

# 7. Définition d'une métrique sur l'espace d'état

Les lois de Newton déterminent l'évolution d'un système.

Le framework GVH cherche à mesurer cette évolution.

Pour cela, il est nécessaire de définir une notion de distance entre deux
états appartenant à l'espace d'état.

Une métrique permet de quantifier le degré de différence entre deux
configurations du système.

Toutes les observables fondamentales du framework GVH seront construites
à partir de cette distance.

## 7.1 Définition d'une métrique

Soit

$begin:math:display$
\\mathcal X\_N
$end:math:display$

l'espace d'état d'un système composé de $begin:math:text$N$end:math:text$ corps.

Une métrique est une application

$begin:math:display$
d \:
\\mathcal X\_N
\\times
\\mathcal X\_N
\\longrightarrow
\\mathbb R\^\{\+\}\,
$end:math:display$

qui associe une distance à toute paire d'états.

Cette métrique doit satisfaire quatre propriétés fondamentales :

1.

$begin:math:display$
d\(X\,Y\)\\ge0
$end:math:display$

2.

$begin:math:display$
d\(X\,Y\)\=0
\\Longleftrightarrow
X\=Y
$end:math:display$

3.

$begin:math:display$
d\(X\,Y\)\=d\(Y\,X\)
$end:math:display$

4.

$begin:math:display$
d\(X\,Z\)
\\le
d\(X\,Y\)\+d\(Y\,Z\)
$end:math:display$

Ces propriétés garantissent une mesure cohérente de la séparation entre
deux états.

## 7.2 Limites de la distance euclidienne classique

Considérons deux états

$begin:math:display$
X\=
\(\\mathbf r\,\\mathbf v\)
$end:math:display$

et

$begin:math:display$
Y\=
\(\\mathbf r\'\,\\mathbf v\'\)\.
$end:math:display$

Une distance euclidienne naïve serait

$begin:math:display$
\\sqrt\{
\\\|\\mathbf r\-\\mathbf r\'\\\|\^2
\+
\\\|\\mathbf v\-\\mathbf v\'\\\|\^2
\}\.
$end:math:display$

Cette expression présente une difficulté fondamentale.

Les positions sont exprimées en mètres.

Les vitesses sont exprimées en mètres par seconde.

Ces deux quantités ne possèdent pas les mêmes dimensions physiques.

Il est donc impossible de les additionner directement sans
normalisation.

Le framework GVH introduit donc une métrique adimensionnée.

## 7.3 Normalisation des variables

Soient

$begin:math:display$
L\_\\star\>0
$end:math:display$

une longueur caractéristique,

et

$begin:math:display$
V\_\\star\>0
$end:math:display$

une vitesse caractéristique.

On définit les variables normalisées

$begin:math:display$
\\widetilde\{\\mathbf r\}
\=
\\frac\{\\mathbf r\}\{L\_\\star\}\,
$end:math:display$

et

$begin:math:display$
\\widetilde\{\\mathbf v\}
\=
\\frac\{\\mathbf v\}\{V\_\\star\}\.
$end:math:display$

L'état normalisé devient

$begin:math:display$
\\widetilde X\=
\(
\\widetilde\{\\mathbf r\}\,
\\widetilde\{\\mathbf v\}
\)\.
$end:math:display$

Toutes les composantes deviennent alors sans dimension.

La comparaison entre deux états est ainsi physiquement cohérente.

## 7.4 Première métrique GVH

Entre deux états normalisés

$begin:math:display$
\\widetilde X\_k
$end:math:display$

et

$begin:math:display$
\\widetilde X\_\{k\+1\}\,
$end:math:display$

on définit

$begin:math:display$
d\_\{\\mathcal X\}
\(
\\widetilde X\_k\,
\\widetilde X\_\{k\+1\}
\)
\=
\\sqrt\{
\\\|
\\Delta\\widetilde\{\\mathbf r\}
\\\|\^2
\+
\\\|
\\Delta\\widetilde\{\\mathbf v\}
\\\|\^2
\}\,
$end:math:display$

où

$begin:math:display$
\\Delta\\widetilde\{\\mathbf r\}
\=
\\widetilde\{\\mathbf r\}\_\{k\+1\}
\-
\\widetilde\{\\mathbf r\}\_k\,
$end:math:display$

et

$begin:math:display$
\\Delta\\widetilde\{\\mathbf v\}
\=
\\widetilde\{\\mathbf v\}\_\{k\+1\}
\-
\\widetilde\{\\mathbf v\}\_k\.
$end:math:display$

Cette métrique constitue la distance fondamentale du framework GVH.

## 7.5 Observable fondamentale

La première observable GVH est définie par

$begin:math:display$
D\_T\(k\)
\=
d\_\{\\mathcal X\}
\(
\\widetilde X\_k\,
\\widetilde X\_\{k\+1\}
\)\.
$end:math:display$

Autrement dit,

$begin:math:display$
D\_T
$end:math:display$

est la distance entre deux états successifs dans l'espace d'état
normalisé.

Toutes les autres observables du framework seront construites à partir de
cette quantité fondamentale.

# 8. Propriétés mathématiques de la métrique GVH

La métrique

$begin:math:display$
d\_\{\\mathcal X\}
\:
\\mathcal X\_N
\\times
\\mathcal X\_N
\\rightarrow
\\mathbb R\^\+
